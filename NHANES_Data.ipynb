```python

# ======================================================================
# NHANES 2005–2018
# NON-GLYCEMIC LABORATORY MARKERS FOR UNDIAGNOSED DIABETES
#
# COMPLETE REPRODUCIBLE ANALYSIS — VERSION 4.0
#
# Single-cell Google Colab / standalone Python script
# Author: To be completed by the researcher
#
# DATA SOURCE: US CDC, National Health and Nutrition Examination Survey
# PERIOD: 2005–2006 through 2017–2018
#
# This script downloads all required data from scratch.
# No previous project files, datasets, models, or outputs are required.
#
# PRIMARY OBJECTIVE:
# Assess whether non-glycemic laboratory measurements improve
# prediction of previously undiagnosed diabetes beyond
# routinely available clinical characteristics.
#
# IMPORTANT SCIENTIFIC RESTRICTIONS:
# 1. This is a predictive, observational study.
# 2. No unverified laboratory conversion equations are applied.
# 3. Assay harmonization is explicitly flagged for review.
# 4. Bootstrap intervals are approximate sensitivity intervals,
#    not formal full NHANES design-based confidence intervals.
# 5. The 2017–2018 cycle was previously explored in this project;
#    it must not be described as a fully untouched holdout.
# 6. No actual monetary laboratory costs are assumed.
# 7. No clinical deployment or diagnostic use is established.
# ======================================================================

import os
import sys
import io
import json
import time
import warnings
import subprocess
import importlib
from pathlib import Path
from datetime import datetime, timezone

def require_package(module, package=None):
    try:
        importlib.import_module(module)
    except ImportError:
        subprocess.check_call([
            sys.executable, "-m", "pip", "install", "-q",
            package or module
        ])

require_package("requests")
require_package("sklearn", "scikit-learn")

import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    log_loss,
    roc_curve,
    precision_recall_curve
)

warnings.filterwarnings("ignore", category=FutureWarning)

# ======================================================================
# 01 — CONFIGURATION
# ======================================================================

SEED = 42
RNG = np.random.default_rng(SEED)

CYCLES = {
    "2005-2006": ("D", 2005),
    "2007-2008": ("E", 2007),
    "2009-2010": ("F", 2009),
    "2011-2012": ("G", 2011),
    "2013-2014": ("H", 2013),
    "2015-2016": ("I", 2015),
    "2017-2018": ("J", 2017)
}

TEST_CYCLE = "2017-2018"

OUTER_FOLDS = 4
INNER_FOLDS = 3
BOOTSTRAP_REPEATS = 1000

REGULARIZATION_GRID = [0.01, 0.1, 1.0]

MAX_SELECTED_LABS = 8

# Minimum CV log-loss reduction required to select a new lab.
# This threshold is a prespecified heuristic, not a clinically
# validated minimal important difference.
MIN_LOGLOSS_IMPROVEMENT = 0.0005

MIN_LAB_CYCLE_COVERAGE = 0.75
MIN_CLINICAL_CYCLE_COVERAGE = 0.70

RUN_NESTED_CV = True

# Source documents must be independently verified before changing.
ASSAY_HARMONIZATION_VERIFIED = False
OUTCOME_HARMONIZATION_VERIFIED = False
FORMAL_SURVEY_VARIANCE_VERIFIED = False
INDEPENDENT_EXTERNAL_VALIDATION = False

# Always fetch files again. Local cache is not used.
FORCE_REDOWNLOAD = True

# ======================================================================
# 02 — OUTPUT DIRECTORIES
# ======================================================================

try:
    from google.colab import drive
    drive.mount("/content/drive")
    ROOT = Path("/content/drive/MyDrive/Sta")
except ImportError:
    ROOT = Path.cwd()

OUT = ROOT / "NHANES_Q2_REPRODUCIBLE_V4"
RAW = OUT / "RAW_CDC_FILES"
TABLES = OUT / "TABLES"
FIGURES = OUT / "FIGURES"
LOGS = OUT / "LOGS"
DATA = OUT / "DATA"

for folder in [OUT, RAW, TABLES, FIGURES, LOGS, DATA]:
    folder.mkdir(parents=True, exist_ok=True)

LOG_MESSAGES = []
DOWNLOAD_RECORDS = []
VARIABLE_RECORDS = []

def log(message):
    line = (
        f"[{datetime.now().strftime('%H:%M:%S')}] "
        f"{message}"
    )
    print(line, flush=True)
    LOG_MESSAGES.append(line)

def save_table(frame, filename):
    frame.to_csv(TABLES / filename, index=False)

def save_plot(filename):
    plt.tight_layout()
    plt.savefig(
        FIGURES / filename,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

log("=" * 76)
log("NHANES Q2 FULL REPRODUCIBLE PIPELINE")
log("=" * 76)
log(f"Output: {OUT}")

# ======================================================================
# 03 — CDC DATA DOWNLOAD
# ======================================================================

session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0 Academic NHANES Research"
})

def get_cdc_file(component, suffix, year, required=False):
    filename = f"{component}_{suffix}.XPT"

    urls = [
        (
            "https://wwwn.cdc.gov/Nchs/Data/Nhanes/"
            f"Public/{year}/DataFiles/{filename}"
        ),
        (
            "https://wwwn.cdc.gov/Nchs/Nhanes/"
            f"{year}-{year+1}/{filename}"
        )
    ]

    errors = []

    for url in urls:
        for attempt in range(3):
            try:
                response = session.get(
                    url,
                    timeout=90
                )
                response.raise_for_status()

                content = response.content

                if len(content) < 100:
                    raise ValueError("Empty or invalid download")

                frame = pd.read_sas(
                    io.BytesIO(content),
                    format="xport"
                )

                if "SEQN" not in [
                    str(x).upper()
                    for x in frame.columns
                ]:
                    raise ValueError("SEQN not found")

                (RAW / filename).write_bytes(content)

                DOWNLOAD_RECORDS.append({
                    "filename": filename,
                    "status": "success",
                    "rows": len(frame),
                    "url": url
                })

                return frame

            except Exception as exc:
                errors.append(str(exc))
                time.sleep(attempt + 1)

    DOWNLOAD_RECORDS.append({
        "filename": filename,
        "status": "failed",
        "rows": 0,
        "url": "; ".join(errors[-3:])
    })

    if required:
        pd.DataFrame(DOWNLOAD_RECORDS).to_csv(
            LOGS / "download_log.csv",
            index=False
        )
        raise RuntimeError(
            f"Required NHANES file unavailable: {filename}\n"
            + "\n".join(errors[-3:])
        )

    log(f"Optional file unavailable: {filename}")
    return None


# ======================================================================
# 04 — NHANES VARIABLE DICTIONARY
# ======================================================================

# Each mapping records actual source variables.
# A missing source variable is never silently fabricated.

VARIABLE_MAP = {
    "DEMO": {
        "age": ["RIDAGEYR"],
        "sex": ["RIAGENDR"],
        "race_ethnicity": ["RIDRETH1"],
        "pregnancy_status": ["RIDEXPRG"],
        "fasting_weight": ["WTSAF2YR"],
        "design_stratum": ["SDMVSTRA"],
        "design_psu": ["SDMVPSU"]
    },
    "DIQ": {
        "diagnosed_diabetes": ["DIQ010"]
    },
    "BMX": {
        "bmi": ["BMXBMI"],
        "waist_circumference": ["BMXWAIST"]
    },
    "GLU": {
        "fasting_glucose": ["LBXGLU"],
        "fasting_weight_fallback": ["WTSAF2YR"]
    },
    "GHB": {
        "hba1c": ["LBXGH"]
    },
    "BIOPRO": {
        "serum_albumin": ["LBXSAL"],
        "serum_creatinine": ["LBXSCR"],
        "uric_acid": ["LBXSUA"],
        "alt": ["LBXSATSI"],
        "ast": ["LBXSASSI"],
        "total_protein": ["LBXSTP"],
        "ggt": ["LBXSGTSI", "LBXSGTS"]
    },
    "CBC": {
        "hemoglobin": ["LBXHGB"],
        "white_blood_cell_count": ["LBXWBCSI"]
    },
    "HDL": {
        "hdl_cholesterol": ["LBDHDD", "LBXHDD"]
    },
    "TCHOL": {
        "total_cholesterol": ["LBXTC"]
    },
    "TRIGLY": {
        "triglycerides": ["LBXTR"]
    },
    "BPX": {
        "sbp1": ["BPXSY1"],
        "sbp2": ["BPXSY2"],
        "sbp3": ["BPXSY3"],
        "sbp4": ["BPXSY4"],
        "dbp1": ["BPXDI1"],
        "dbp2": ["BPXDI2"],
        "dbp3": ["BPXDI3"],
        "dbp4": ["BPXDI4"]
    },
    "SMQ": {
        "smoking_ever": ["SMQ020"]
    },
    "MCQ": {
        "family_history_diabetes": ["MCQ300C"]
    },
    "PAQ": {
        "vigorous_work": ["PAQ605"],
        "moderate_work": ["PAQ620"],
        "vigorous_recreation": ["PAQ650"],
        "moderate_recreation": ["PAQ665"],
        "old_vigorous_activity": ["PAD200"],
        "old_moderate_activity": ["PAD320"]
    }
}

REQUIRED_COMPONENTS = {
    "DEMO", "DIQ", "BMX", "GLU", "GHB"
}

ALL_COMPONENTS = list(VARIABLE_MAP.keys())

def extract_variables(frame, component, cycle):
    if frame is None:
        return None

    frame = frame.copy()
    frame.columns = [
        str(c).upper()
        for c in frame.columns
    ]

    selected = ["SEQN"]
    rename = {}

    for target, candidates in VARIABLE_MAP[component].items():
        matched = next(
            (
                candidate
                for candidate in candidates
                if candidate in frame.columns
            ),
            None
        )

        VARIABLE_RECORDS.append({
            "cycle": cycle,
            "component": component,
            "target_variable": target,
            "source_variable": matched,
            "found": matched is not None
        })

        if matched is not None:
            selected.append(matched)
            rename[matched] = target

    selected = list(dict.fromkeys(selected))

    result = frame[selected].rename(
        columns=rename
    )

    if result["SEQN"].duplicated().any():
        raise ValueError(
            f"Duplicate SEQN in {component}, {cycle}"
        )

    return result


# ======================================================================
# 05 — REBUILD COMPLETE COHORT
# ======================================================================

CORE_CLINICAL = [
    "age",
    "sex",
    "bmi",
    "waist_circumference"
]

OPTIONAL_CLINICAL = [
    "systolic_bp",
    "diastolic_bp",
    "smoking_ever",
    "family_history_diabetes"
]

LABORATORIES = [
    "serum_albumin",
    "serum_creatinine",
    "uric_acid",
    "alt",
    "ast",
    "total_protein",
    "total_cholesterol",
    "hdl_cholesterol",
    "hemoglobin",
    "white_blood_cell_count",
    "triglycerides",
    "ggt"
]

cycle_frames = []
cohort_rows = []

for cycle, (suffix, year) in CYCLES.items():
    log(f"REBUILDING CYCLE: {cycle}")

    components = {}

    for component in ALL_COMPONENTS:
        raw = get_cdc_file(
            component,
            suffix,
            year,
            required=component in REQUIRED_COMPONENTS
        )

        components[component] = extract_variables(
            raw, component, cycle
        )

    frame = components["DEMO"].copy()

    for component in ALL_COMPONENTS:
        if component == "DEMO":
            continue

        piece = components[component]

        if piece is not None:
            frame = frame.merge(
                piece,
                on="SEQN",
                how="left",
                validate="one_to_one"
            )

    frame["cycle"] = cycle

    for column in frame.columns:
        if column != "cycle":
            frame[column] = pd.to_numeric(
                frame[column],
                errors="coerce"
            )

    if "fasting_weight" not in frame.columns:
        frame["fasting_weight"] = np.nan

    if "fasting_weight_fallback" in frame.columns:
        frame["fasting_weight"] = (
            frame["fasting_weight"].fillna(
                frame["fasting_weight_fallback"]
            )
        )

    # Verify mandatory outcome and design variables.
    mandatory = [
        "age", "sex",
        "diagnosed_diabetes",
        "fasting_glucose",
        "hba1c",
        "fasting_weight",
        "design_stratum",
        "design_psu"
    ]

    missing_mandatory = [
        x for x in mandatory
        if x not in frame.columns
    ]

    if missing_mandatory:
        raise RuntimeError(
            f"{cycle}: missing mandatory variables "
            f"{missing_mandatory}"
        )

    for column in (
        CORE_CLINICAL +
        OPTIONAL_CLINICAL +
        LABORATORIES
    ):
        if column not in frame.columns:
            frame[column] = np.nan

    # Blood pressure: mean of available positive readings.
    for prefix, output in [
        ("sbp", "systolic_bp"),
        ("dbp", "diastolic_bp")
    ]:
        cols = [
            f"{prefix}{i}"
            for i in range(1, 5)
            if f"{prefix}{i}" in frame.columns
        ]

        if cols:
            frame[output] = (
                frame[cols]
                .where(frame[cols] > 0)
                .mean(axis=1)
            )

    # Valid questionnaire responses only.
    for column in [
        "smoking_ever",
        "family_history_diabetes"
    ]:
        frame[column] = frame[column].map({
            1: 1.0,
            2: 0.0
        })

    # Physical activity questionnaires changed across cycles.
    # We do not pool the 2005 questions with later questions.
    # Activity is excluded from the primary pooled model.
    # The original responses remain available for audit.

    # Exclusion flags are retained for reproducibility.
    frame["exclude_age"] = ~frame["age"].ge(20)

    # DIQ010 == 2: respondent reports no diabetes diagnosis.
    frame["exclude_diagnosis"] = (
        ~frame["diagnosed_diabetes"].eq(2)
    )

    if "pregnancy_status" in frame.columns:
        frame["exclude_pregnancy"] = (
            frame["pregnancy_status"].eq(1)
        )
    else:
        frame["exclude_pregnancy"] = False

    frame["exclude_missing_outcome"] = (
        frame["hba1c"].isna() |
        frame["fasting_glucose"].isna()
    )

    frame["exclude_weight"] = (
        ~frame["fasting_weight"].gt(0)
    )

    frame["exclude_design"] = (
        frame["design_stratum"].isna() |
        frame["design_psu"].isna()
    )

    exclude_columns = [
        "exclude_age",
        "exclude_diagnosis",
        "exclude_pregnancy",
        "exclude_missing_outcome",
        "exclude_weight",
        "exclude_design"
    ]

    eligible = ~frame[exclude_columns].any(axis=1)

    included = frame.loc[eligible].copy()

    included["outcome"] = (
        included["hba1c"].ge(6.5) |
        included["fasting_glucose"].ge(126)
    ).astype(int)

    # Seven two-year cycles: combined 14-year subsample weight.
    included["weight"] = (
        included["fasting_weight"] / 7.0
    )

    included["participant_id"] = (
        cycle + "_" +
        included["SEQN"].astype(int).astype(str)
    )

    included["psu_group"] = (
        cycle + "_" +
        included["design_stratum"].astype(int).astype(str)
        + "_" +
        included["design_psu"].astype(int).astype(str)
    )

    included["stratum_group"] = (
        cycle + "_" +
        included["design_stratum"].astype(int).astype(str)
    )

    cohort_rows.append({
        "cycle": cycle,
        "initial_n": len(frame),
        "excluded_age": int(
            frame["exclude_age"].sum()
        ),
        "excluded_diagnosis": int(
            frame["exclude_diagnosis"].sum()
        ),
        "excluded_pregnancy": int(
            frame["exclude_pregnancy"].sum()
        ),
        "excluded_missing_outcome": int(
            frame["exclude_missing_outcome"].sum()
        ),
        "excluded_invalid_weight": int(
            frame["exclude_weight"].sum()
        ),
        "eligible_n": len(included),
        "cases": int(included["outcome"].sum()),
        "weighted_prevalence": (
            np.average(
                included["outcome"],
                weights=included["weight"]
            )
            if len(included) else np.nan
        ),
        "unique_stratum_psu_pairs": (
            included["psu_group"].nunique()
        )
    })

    if len(included) == 0:
        raise RuntimeError(
            f"No eligible observations in {cycle}"
        )

    cycle_frames.append(included)

df = pd.concat(
    cycle_frames,
    ignore_index=True,
    sort=False
)

if not df["participant_id"].is_unique:
    raise ValueError("Duplicate participant IDs")

if not df["weight"].gt(0).all():
    raise ValueError("Invalid survey weights")

save_table(
    pd.DataFrame(cohort_rows),
    "Table_01_Cohort.csv"
)

save_table(
    pd.DataFrame(VARIABLE_RECORDS),
    "Supplement_Source_Variable_Mapping.csv"
)

pd.DataFrame(DOWNLOAD_RECORDS).to_csv(
    LOGS / "CDC_Download_Log.csv",
    index=False
)

df.to_csv(
    DATA / "NHANES_Rebuilt_Cohort.csv",
    index=False
)

log(f"TOTAL COHORT N={len(df)}")
log(f"TOTAL CASES={int(df['outcome'].sum())}")

# ======================================================================
# 06 — DEVELOPMENT / TEMPORAL EVALUATION
# ======================================================================

dev = df.loc[
    df["cycle"] != TEST_CYCLE
].copy().reset_index(drop=True)

test = df.loc[
    df["cycle"] == TEST_CYCLE
].copy().reset_index(drop=True)

log(f"Development N={len(dev)}")
log(f"Temporal evaluation N={len(test)}")

# ======================================================================
# 07 — FEATURE AVAILABILITY
# ======================================================================

coverage_rows = []

for feature in (
    CORE_CLINICAL +
    OPTIONAL_CLINICAL +
    LABORATORIES +
    ["race_ethnicity"]
):
    for cycle, group in df.groupby("cycle"):
        coverage_rows.append({
            "feature": feature,
            "cycle": cycle,
            "coverage": group[feature].notna().mean()
                if feature in group.columns else 0,
            "median": group[feature].median()
                if feature in group.columns else np.nan
        })

coverage = pd.DataFrame(coverage_rows)

save_table(
    coverage,
    "Table_02_Feature_Coverage.csv"
)

minimum_coverage = (
    coverage.groupby("feature")["coverage"]
    .min()
    .to_dict()
)

missing_core = [
    x for x in CORE_CLINICAL
    if minimum_coverage.get(x, 0) <
    MIN_CLINICAL_CYCLE_COVERAGE
]

if missing_core:
    raise RuntimeError(
        "Insufficient coverage for mandatory predictors: "
        + str(missing_core)
    )

available_optional = [
    x for x in OPTIONAL_CLINICAL
    if minimum_coverage.get(x, 0) >=
    MIN_CLINICAL_CYCLE_COVERAGE
]

available_labs = [
    x for x in LABORATORIES
    if minimum_coverage.get(x, 0) >=
    MIN_LAB_CYCLE_COVERAGE
]

# Preserve the original 12-lab candidate definition.
# Exclusions must be reported explicitly.
excluded_labs = [
    x for x in LABORATORIES
    if x not in available_labs
]

# One-hot encode race/ethnicity, using the same fixed coding
# for development and temporal evaluation.
race_columns = []

if minimum_coverage.get("race_ethnicity", 0) >= 0.70:
    for category in [2, 3, 4, 5]:
        column = f"race_category_{category}"
        race_columns.append(column)

        for frame in [dev, test]:
            frame[column] = np.where(
                frame["race_ethnicity"].notna(),
                (
                    frame["race_ethnicity"] == category
                ).astype(float),
                np.nan
            )

CLINICAL = (
    CORE_CLINICAL +
    available_optional +
    race_columns
)

if len(available_labs) < 3:
    raise RuntimeError(
        "Insufficient laboratory coverage; "
        "review source mapping."
    )

log(f"CLINICAL FEATURES: {CLINICAL}")
log(f"AVAILABLE LABORATORIES: {available_labs}")
log(f"EXCLUDED LABORATORIES: {excluded_labs}")

save_table(
    pd.DataFrame([
        {
            "feature": x,
            "minimum_cycle_coverage":
                minimum_coverage.get(x, np.nan),
            "included_clinical": x in CLINICAL,
            "included_laboratory": x in available_labs
        }
        for x in (
            CORE_CLINICAL +
            OPTIONAL_CLINICAL +
            LABORATORIES
        )
    ]),
    "Table_03_Feature_Eligibility.csv"
)

# ======================================================================
# 08 — MODEL UTILITIES
# ======================================================================

def sample_weights(frame):
    w = frame["weight"].to_numpy(dtype=float)
    return w / np.mean(w)

def model_pipeline(C):
    return Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                keep_empty_features=True
            )
        ),
        ("scaler", StandardScaler()),
        (
            "model",
            LogisticRegression(
                C=C,
                max_iter=3000,
                random_state=SEED
            )
        )
    ])

def fit_model(frame, features, C):
    model = model_pipeline(C)
    model.fit(
        frame[features],
        frame["outcome"],
        model__sample_weight=sample_weights(frame)
    )
    return model

def predict(model, frame, features):
    return model.predict_proba(
        frame[features]
    )[:, 1]

def evaluate(y, p, w):
    y = np.asarray(y, dtype=int)
    p = np.clip(
        np.asarray(p, dtype=float),
        1e-8, 1 - 1e-8
    )
    w = np.asarray(w, dtype=float)

    result = {
        "n": len(y),
        "cases": int(y.sum()),
        "weighted_prevalence": float(
            np.average(y, weights=w)
        ),
        "mean_predicted_risk": float(
            np.average(p, weights=w)
        ),
        "brier": float(
            brier_score_loss(
                y, p, sample_weight=w
            )
        ),
        "log_loss": float(
            log_loss(
                y, p,
                sample_weight=w,
                labels=[0, 1]
            )
        ),
        "auroc": np.nan,
        "auprc": np.nan
    }

    if np.unique(y).size == 2:
        result["auroc"] = float(
            roc_auc_score(
                y, p, sample_weight=w
            )
        )
        result["auprc"] = float(
            average_precision_score(
                y, p, sample_weight=w
            )
        )

    return result

def folds_for(frame, k, seed):
    splitter = StratifiedGroupKFold(
        n_splits=k,
        shuffle=True,
        random_state=seed
    )

    return list(
        splitter.split(
            frame,
            frame["outcome"],
            frame["psu_group"]
        )
    )

def cv_losses(frame, features, C, folds):
    losses = []

    for train_idx, valid_idx in folds:
        train = frame.iloc[train_idx]
        valid = frame.iloc[valid_idx]

        if (
            train["outcome"].nunique() < 2 or
            valid["outcome"].nunique() < 2
        ):
            return np.array([np.inf])

        model = fit_model(
            train, features, C
        )

        p = predict(
            model, valid, features
        )

        losses.append(
            log_loss(
                valid["outcome"],
                p,
                sample_weight=valid["weight"],
                labels=[0, 1]
            )
        )

    return np.asarray(losses)

def tune(frame, features, folds):
    results = []

    for C in REGULARIZATION_GRID:
        losses = cv_losses(
            frame, features, C, folds
        )

        results.append({
            "C": C,
            "mean_loss": float(np.mean(losses)),
            "fold_losses": losses
        })

    return min(
        results,
        key=lambda x: x["mean_loss"]
    )

# ======================================================================
# 09 — SEQUENTIAL LABORATORY SELECTION
# ======================================================================

def select_laboratories(
    frame, clinical, candidates, seed
):
    folds = folds_for(
        frame,
        INNER_FOLDS,
        seed
    )

    baseline = tune(
        frame, clinical, folds
    )

    selected = []
    history = []

    current_loss = baseline["mean_loss"]

    for step in range(MAX_SELECTED_LABS):
        remaining = [
            x for x in candidates
            if x not in selected
        ]

        if not remaining:
            break

        options = []

        for lab in remaining:
            features = (
                clinical + selected + [lab]
            )

            candidate = tune(
                frame, features, folds
            )

            improvement = (
                current_loss -
                candidate["mean_loss"]
            )

            options.append({
                "laboratory": lab,
                "mean_loss": candidate["mean_loss"],
                "improvement": improvement,
                "C": candidate["C"]
            })

        best = max(
            options,
            key=lambda x: x["improvement"]
        )

        if (
            best["improvement"] <
            MIN_LOGLOSS_IMPROVEMENT
        ):
            break

        selected.append(
            best["laboratory"]
        )

        current_loss = best["mean_loss"]

        history.append({
            "step": step + 1,
            **best
        })

    final = tune(
        frame,
        clinical + selected,
        folds
    )

    return {
        "selected": selected,
        "history": history,
        "C": final["C"],
        "cv_loss": final["mean_loss"]
    }

# ======================================================================
# 10 — GROUP-AWARE NESTED CROSS-VALIDATION
# ======================================================================

nested_rows = []
selection_rows = []

if RUN_NESTED_CV:
    log("STARTING NESTED CROSS-VALIDATION")

    outer_folds = folds_for(
        dev, OUTER_FOLDS, SEED
    )

    for fold_number, (tr, va) in enumerate(
        outer_folds, 1
    ):
        log(
            f"OUTER FOLD {fold_number}/{OUTER_FOLDS}"
        )

        outer_train = dev.iloc[tr].copy()
        outer_valid = dev.iloc[va].copy()

        selection = select_laboratories(
            outer_train,
            CLINICAL,
            available_labs,
            SEED + fold_number
        )

        for lab in selection["selected"]:
            selection_rows.append({
                "fold": fold_number,
                "laboratory": lab
            })

        inner = folds_for(
            outer_train,
            INNER_FOLDS,
            SEED + 100 + fold_number
        )

        definitions = {
            "clinical": CLINICAL,
            "selected": (
                CLINICAL + selection["selected"]
            ),
            "full": (
                CLINICAL + available_labs
            )
        }

        for name, features in definitions.items():
            tuned = tune(
                outer_train,
                features,
                inner
            )

            model = fit_model(
                outer_train,
                features,
                tuned["C"]
            )

            p = predict(
                model,
                outer_valid,
                features
            )

            nested_rows.append({
                "fold": fold_number,
                "model": name,
                "C": tuned["C"],
                "selected_labs": ",".join(
                    selection["selected"]
                ),
                **evaluate(
                    outer_valid["outcome"],
                    p,
                    outer_valid["weight"]
                )
            })

    nested = pd.DataFrame(nested_rows)

    save_table(
        nested,
        "Table_04_Nested_CV.csv"
    )

    frequency = (
        pd.DataFrame(selection_rows)
        .groupby("laboratory")
        .size()
        .reset_index(name="selected_folds")
    )

    frequency["frequency"] = (
        frequency["selected_folds"] /
        OUTER_FOLDS
    )

    save_table(
        frequency,
        "Table_05_Selection_Stability.csv"
    )

    log("NESTED CV COMPLETE")
    display(nested.round(5))
    display(frequency)

# ======================================================================
# 11 — INDEPENDENT CALIBRATION PARTITION
# ======================================================================

log("CREATING GROUP-SEPARATED CALIBRATION DATA")

# A group-aware split is made before final variable selection.
# No calibration participant contributes to selection or tuning.

splitter = StratifiedGroupKFold(
    n_splits=4,
    shuffle=True,
    random_state=SEED + 500
)

train_idx, cal_idx = next(
    splitter.split(
        dev,
        dev["outcome"],
        dev["psu_group"]
    )
)

train = dev.iloc[
    train_idx
].copy().reset_index(drop=True)

cal = dev.iloc[
    cal_idx
].copy().reset_index(drop=True)

assert set(train["psu_group"]).isdisjoint(
    set(cal["psu_group"])
)

if (
    train["outcome"].nunique() < 2 or
    cal["outcome"].nunique() < 2
):
    raise RuntimeError(
        "Training/calibration split lacks outcome classes."
    )

log(f"TRAIN N={len(train)}")
log(f"CALIBRATION N={len(cal)}")

# ======================================================================
# 12 — FINAL LABORATORY SELECTION
# ======================================================================

selection = select_laboratories(
    train,
    CLINICAL,
    available_labs,
    SEED + 700
)

selected_labs = selection["selected"]

save_table(
    pd.DataFrame(selection["history"]),
    "Table_06_Final_Laboratory_Selection.csv"
)

log(f"SELECTED LABORATORIES: {selected_labs}")

# ======================================================================
# 13 — MODEL TRAINING AND CALIBRATION
# ======================================================================

definitions = {
    "clinical": CLINICAL,
    "selected": (
        CLINICAL + selected_labs
    ),
    "full": (
        CLINICAL + available_labs
    )
}

final_predictions = {}
final_regularization = {}

inner = folds_for(
    train,
    INNER_FOLDS,
    SEED + 900
)

def logit_transform(p):
    p = np.clip(
        np.asarray(p),
        1e-6, 1 - 1e-6
    )
    return np.log(
        p / (1 - p)
    ).reshape(-1, 1)

for name, features in definitions.items():
    log(f"FITTING MODEL: {name}")

    tuned = tune(
        train,
        features,
        inner
    )

    model = fit_model(
        train,
        features,
        tuned["C"]
    )

    p_cal = predict(
        model, cal, features
    )

    p_test = predict(
        model, test, features
    )

    recalibrator = LogisticRegression(
        C=1e6,
        max_iter=3000,
        random_state=SEED
    )

    recalibrator.fit(
        logit_transform(p_cal),
        cal["outcome"],
        sample_weight=sample_weights(cal)
    )

    p_final = recalibrator.predict_proba(
        logit_transform(p_test)
    )[:, 1]

    final_predictions[name] = p_final
    final_regularization[name] = tuned["C"]

# ======================================================================
# 14 — TEMPORAL PERFORMANCE
# ======================================================================

y = test["outcome"].to_numpy(dtype=int)
w = test["weight"].to_numpy(dtype=float)

performance = pd.DataFrame([
    {
        "model": name,
        "C": final_regularization[name],
        "laboratory_count": (
            len(definitions[name]) -
            len(CLINICAL)
        ),
        **evaluate(y, p, w)
    }
    for name, p in final_predictions.items()
])

save_table(
    performance,
    "Table_07_Temporal_Performance.csv"
)

log("TEMPORAL PERFORMANCE")
display(performance.round(5))

# ======================================================================
# 15 — PAIRED COMPARISONS
# ======================================================================

comparisons = [
    ("selected_vs_clinical", "selected", "clinical"),
    ("full_vs_clinical", "full", "clinical"),
    ("selected_vs_full", "selected", "full")
]

comparison_rows = []

for label, a, b in comparisons:
    ma = evaluate(y, final_predictions[a], w)
    mb = evaluate(y, final_predictions[b], w)

    comparison_rows.append({
        "comparison": label,
        **{
            f"delta_{metric}": (
                ma[metric] - mb[metric]
            )
            for metric in [
                "auroc", "auprc",
                "brier", "log_loss"
            ]
        }
    })

comparison_table = pd.DataFrame(
    comparison_rows
)

save_table(
    comparison_table,
    "Table_08_Model_Differences.csv"
)

display(comparison_table.round(5))

# ======================================================================
# 16 — STRATIFIED PSU BOOTSTRAP
# ======================================================================

log("RUNNING APPROXIMATE PSU BOOTSTRAP")

strata = test["stratum_group"].to_numpy()
psus = test["psu_group"].to_numpy()

clusters = {}

for stratum in np.unique(strata):
    positions = np.where(
        strata == stratum
    )[0]

    clusters[stratum] = [
        positions[psus[positions] == psu]
        for psu in np.unique(psus[positions])
    ]

bootstrap_rows = []

for repeat in range(BOOTSTRAP_REPEATS):
    sampled = []

    for stratum, units in clusters.items():
        choices = RNG.integers(
            0,
            len(units),
            size=len(units)
        )

        for choice in choices:
            sampled.extend(
                units[choice].tolist()
            )

    idx = np.asarray(
        sampled, dtype=int
    )

    if np.unique(y[idx]).size < 2:
        continue

    replicate_metrics = {
        name: evaluate(
            y[idx],
            p[idx],
            w[idx]
        )
        for name, p in final_predictions.items()
    }

    for label, a, b in comparisons:
        for metric in [
            "auroc", "auprc",
            "brier", "log_loss"
        ]:
            bootstrap_rows.append({
                "repeat": repeat,
                "comparison": label,
                "metric": metric,
                "difference": (
                    replicate_metrics[a][metric] -
                    replicate_metrics[b][metric]
                )
            })

bootstrap = pd.DataFrame(
    bootstrap_rows
)

bootstrap.to_csv(
    DATA / "Bootstrap_Replicates.csv",
    index=False
)

intervals = (
    bootstrap.groupby(
        ["comparison", "metric"]
    )["difference"]
    .agg(
        bootstrap_mean="mean",
        lower_95=lambda x: x.quantile(0.025),
        upper_95=lambda x: x.quantile(0.975),
        valid_replicates="count"
    )
    .reset_index()
)

save_table(
    intervals,
    "Table_09_Approximate_Bootstrap_Intervals.csv"
)

display(intervals.round(5))

# ======================================================================
# 17 — CALIBRATION
# ======================================================================

calibration_rows = []
calibration_bins = []

for name, p in final_predictions.items():
    recalibration_model = LogisticRegression(
        C=1e6,
        max_iter=3000,
        random_state=SEED
    )

    recalibration_model.fit(
        logit_transform(p),
        y,
        sample_weight=w / np.mean(w)
    )

    calibration_rows.append({
        "model": name,
        "observed_prevalence": np.average(
            y, weights=w
        ),
        "mean_predicted_risk": np.average(
            p, weights=w
        ),
        "calibration_intercept": (
            recalibration_model.intercept_[0]
        ),
        "calibration_slope": (
            recalibration_model.coef_[0, 0]
        )
    })

    bins = pd.qcut(
        pd.Series(p).rank(method="first"),
        10,
        labels=False,
        duplicates="drop"
    ).to_numpy()

    for bin_id in np.unique(bins):
        mask = bins == bin_id

        calibration_bins.append({
            "model": name,
            "bin": int(bin_id) + 1,
            "n": int(mask.sum()),
            "predicted": np.average(
                p[mask],
                weights=w[mask]
            ),
            "observed": np.average(
                y[mask],
                weights=w[mask]
            )
        })

calibration = pd.DataFrame(
    calibration_rows
)

calibration_curve = pd.DataFrame(
    calibration_bins
)

save_table(
    calibration,
    "Table_10_Calibration.csv"
)

save_table(
    calibration_curve,
    "Supplement_Calibration_Bins.csv"
)

display(calibration.round(5))

# ======================================================================
# 18 — SUBGROUPS
# ======================================================================

test["age_group"] = pd.cut(
    test["age"],
    bins=[19, 39, 59, np.inf],
    labels=["20-39", "40-59", "60+"]
)

test["sex_group"] = test["sex"].map({
    1: "Male",
    2: "Female"
})

subgroup_rows = []

for variable in ["age_group", "sex_group"]:
    values = test[variable].astype(str).to_numpy()

    for group in sorted(np.unique(values)):
        if group == "nan":
            continue

        mask = values == group

        for name, p in final_predictions.items():
            subgroup_rows.append({
                "variable": variable,
                "group": group,
                "model": name,
                **evaluate(
                    y[mask],
                    p[mask],
                    w[mask]
                )
            })

subgroups = pd.DataFrame(
    subgroup_rows
)

save_table(
    subgroups,
    "Table_11_Subgroups.csv"
)

display(subgroups.round(5))

# ======================================================================
# 19 — COMMON COMPLETE-CASE SENSITIVITY
# ======================================================================

all_model_features = list(
    dict.fromkeys(
        CLINICAL + available_labs
    )
)

complete_mask = (
    test[all_model_features]
    .notna()
    .all(axis=1)
    .to_numpy()
)

complete_rows = []

for name, p in final_predictions.items():
    complete_rows.append({
        "model": name,
        "complete_case_fraction": (
            complete_mask.mean()
        ),
        **evaluate(
            y[complete_mask],
            p[complete_mask],
            w[complete_mask]
        )
    })

save_table(
    pd.DataFrame(complete_rows),
    "Table_12_Common_Complete_Case.csv"
)

# ======================================================================
# 20 — DECISION CURVE ANALYSIS
# ======================================================================

thresholds = np.linspace(
    0.01, 0.15, 29
)

weighted_prevalence = np.average(
    y, weights=w
)

wn = w / w.sum()

dca_rows = []

for threshold in thresholds:
    dca_rows.append({
        "threshold": threshold,
        "strategy": "treat_none",
        "net_benefit": 0.0
    })

    dca_rows.append({
        "threshold": threshold,
        "strategy": "treat_all",
        "net_benefit": (
            weighted_prevalence -
            (1 - weighted_prevalence) *
            threshold / (1 - threshold)
        )
    })

    for name, p in final_predictions.items():
        predicted_positive = p >= threshold

        tp = np.sum(
            wn * (
                (y == 1) &
                predicted_positive
            )
        )

        fp = np.sum(
            wn * (
                (y == 0) &
                predicted_positive
            )
        )

        dca_rows.append({
            "threshold": threshold,
            "strategy": name,
            "net_benefit": (
                tp -
                fp * threshold / (1 - threshold)
            )
        })

dca = pd.DataFrame(dca_rows)

save_table(
    dca,
    "Table_13_Decision_Curve.csv"
)

# ======================================================================
# 21 — PUBLICATION FIGURES
# ======================================================================

log("CREATING PUBLICATION FIGURES")

plt.figure(figsize=(7, 6))

for name, p in final_predictions.items():
    fpr, tpr, _ = roc_curve(
        y, p, sample_weight=w
    )

    auc = roc_auc_score(
        y, p, sample_weight=w
    )

    plt.plot(
        fpr, tpr,
        linewidth=2,
        label=f"{name}: AUROC {auc:.3f}"
    )

plt.plot(
    [0, 1], [0, 1],
    linestyle="--",
    linewidth=1
)

plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("Temporal Validation: ROC")
plt.legend()
save_plot("Figure_01_ROC.png")

plt.figure(figsize=(7, 6))

for name, p in final_predictions.items():
    precision, recall, _ = precision_recall_curve(
        y, p, sample_weight=w
    )

    ap = average_precision_score(
        y, p, sample_weight=w
    )

    plt.plot(
        recall, precision,
        linewidth=2,
        label=f"{name}: AUPRC {ap:.3f}"
    )

plt.axhline(
    weighted_prevalence,
    linestyle="--",
    linewidth=1
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Temporal Validation: Precision-Recall")
plt.legend()
save_plot("Figure_02_Precision_Recall.png")

plt.figure(figsize=(7, 6))

for name, group in calibration_curve.groupby("model"):
    plt.plot(
        group["predicted"],
        group["observed"],
        marker="o",
        linewidth=2,
        label=name
    )

plt.plot(
    [0, 0.35],
    [0, 0.35],
    linestyle="--",
    linewidth=1
)

plt.xlabel("Predicted risk")
plt.ylabel("Observed risk")
plt.title("Temporal Calibration")
plt.legend()
save_plot("Figure_03_Calibration.png")

plt.figure(figsize=(8, 5))

for name, group in dca.groupby("strategy"):
    plt.plot(
        group["threshold"],
        group["net_benefit"],
        linewidth=2,
        label=name
    )

plt.axhline(
    0, linestyle=":", linewidth=1
)

plt.xlabel("Risk threshold")
plt.ylabel("Weighted net benefit")
plt.title("Exploratory Decision Curve Analysis")
plt.legend(fontsize=8)
save_plot("Figure_04_Decision_Curve.png")

# ======================================================================
# 22 — PREDICTIONS AND REPRODUCIBILITY
# ======================================================================

prediction_output = test[
    [
        "participant_id",
        "cycle",
        "SEQN",
        "outcome",
        "weight",
        "design_stratum",
        "design_psu"
    ]
].copy()

for name, p in final_predictions.items():
    prediction_output[
        f"probability_{name}"
    ] = p

prediction_output.to_csv(
    DATA / "Temporal_Predictions.csv",
    index=False
)

# ======================================================================
# 23 — SCIENTIFIC QUALITY GATES
# ======================================================================

quality = pd.DataFrame([
    {
        "check": "All seven cycles available",
        "passed": df["cycle"].nunique() == 7
    },
    {
        "check": "Unique participants",
        "passed": df["participant_id"].is_unique
    },
    {
        "check": "Positive fasting weights",
        "passed": df["weight"].gt(0).all()
    },
    {
        "check": "Outcome markers excluded from predictors",
        "passed": not any(
            x in (
                CLINICAL + available_labs
            )
            for x in [
                "hba1c", "fasting_glucose"
            ]
        )
    },
    {
        "check": "Calibration separated before selection",
        "passed": set(
            train["participant_id"]
        ).isdisjoint(
            set(cal["participant_id"])
        )
    },
    {
        "check": "PSU-group separation",
        "passed": set(
            train["psu_group"]
        ).isdisjoint(
            set(cal["psu_group"])
        )
    },
    {
        "check": "Nested CV completed",
        "passed": RUN_NESTED_CV
    },
    {
        "check": "Calibration reported",
        "passed": len(calibration) == 3
    },
    {
        "check": "Outcome assay harmonization verified",
        "passed": OUTCOME_HARMONIZATION_VERIFIED
    },
    {
        "check": "Predictor assay harmonization verified",
        "passed": ASSAY_HARMONIZATION_VERIFIED
    },
    {
        "check": "Formal survey-design variance completed",
        "passed": FORMAL_SURVEY_VARIANCE_VERIFIED
    },
    {
        "check": "Independent external validation",
        "passed": INDEPENDENT_EXTERNAL_VALIDATION
    },
    {
        "check": "Temporal evaluation fully untouched",
        "passed": False
    }
])

save_table(
    quality,
    "Scientific_Quality_Gates.csv"
)

# ======================================================================
# 24 — SAVE ANALYSIS MANIFEST
# ======================================================================

manifest = {
    "study": (
        "Incremental Predictive Value of Non-Glycemic "
        "Laboratory Markers for Undiagnosed Diabetes"
    ),
    "created_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "seed": SEED,
    "cycles": list(CYCLES.keys()),
    "development_cycles": [
        x for x in CYCLES
        if x != TEST_CYCLE
    ],
    "temporal_cycle": TEST_CYCLE,
    "total_n": len(df),
    "total_cases": int(df["outcome"].sum()),
    "development_n": len(dev),
    "temporal_n": len(test),
    "temporal_cases": int(test["outcome"].sum()),
    "clinical_features": CLINICAL,
    "candidate_laboratories": LABORATORIES,
    "eligible_laboratories": available_labs,
    "excluded_laboratories": excluded_labs,
    "selected_laboratories": selected_labs,
    "regularization": final_regularization,
    "selection_minimum_improvement":
        MIN_LOGLOSS_IMPROVEMENT,
    "nested_outer_folds": OUTER_FOLDS,
    "nested_inner_folds": INNER_FOLDS,
    "bootstrap_repeats": BOOTSTRAP_REPEATS,
    "survey_inference": (
        "Approximate fixed-prediction PSU bootstrap; "
        "not formal full design-based variance"
    ),
    "assay_harmonization_verified":
        ASSAY_HARMONIZATION_VERIFIED,
    "outcome_harmonization_verified":
        OUTCOME_HARMONIZATION_VERIFIED,
    "external_validation":
        INDEPENDENT_EXTERNAL_VALIDATION
}

with open(
    OUT / "Reproducibility_Manifest.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        manifest,
        f,
        indent=2,
        ensure_ascii=False
    )

# ======================================================================
# 25 — SOURCE DOCUMENTATION LINKS
# ======================================================================

documentation_rows = []

for cycle, (suffix, year) in CYCLES.items():
    for component in ALL_COMPONENTS:
        documentation_rows.append({
            "cycle": cycle,
            "component": component,
            "documentation_url": (
                "https://wwwn.cdc.gov/Nchs/Data/Nhanes/"
                f"Public/{year}/DataFiles/"
                f"{component}_{suffix}.htm"
            ),
            "analytic_notes_reviewed": False,
            "assay_comparability_verified": False,
            "adjustment_applied": False
        })

save_table(
    pd.DataFrame(documentation_rows),
    "Supplement_CDC_Documentation_Audit.csv"
)

# ======================================================================
# 26 — FINAL RESULTS AND LOGS
# ======================================================================

log("=" * 76)
log("FINAL ANALYSIS RESULTS")
log("=" * 76)

print("\nCOHORT")
display(pd.DataFrame(cohort_rows).round(5))

print("\nFEATURES")
display(
    pd.read_csv(
        TABLES / "Table_03_Feature_Eligibility.csv"
    )
)

print("\nSELECTED LABORATORIES")
print(selected_labs)

print("\nTEMPORAL PERFORMANCE")
display(performance.round(5))

print("\nMODEL DIFFERENCES")
display(comparison_table.round(5))

print("\nBOOTSTRAP INTERVALS")
display(intervals.round(5))

print("\nCALIBRATION")
display(calibration.round(5))

print("\nSUBGROUPS")
display(subgroups.round(5))

print("\nSCIENTIFIC QUALITY GATES")
display(quality)

log(f"ALL RESULTS SAVED TO: {OUT}")

if not quality["passed"].all():
    log(
        "COMPUTATIONAL ANALYSIS COMPLETED. "
        "SCIENTIFIC SUBMISSION REQUIREMENTS REMAIN."
    )

log("END OF REPRODUCIBLE PIPELINE")

(LOGS / "Execution_Log.txt").write_text(
    "\n".join(LOG_MESSAGES),
    encoding="utf-8"
)
```

```python

# =====================================================================
# NHANES Q2 — FINAL STATISTICAL REPORTING ADDENDUM (V4.2)
# Run AFTER V4 and V4.1 in the SAME Colab session.
#
# No CDC re-downloads.
# No new model fitting or feature selection.
# No reuse of the temporal test for model optimization.
#
# PRIMARY:
#   Full 12-laboratory model vs clinical model
#
# SENSITIVITY:
#   Exclude 2005–2006 from development
#
# SECONDARY:
#   Selected panel and its stability
#
# IMPORTANT:
#   Bootstrap intervals condition on the fitted models.
#   They do not account for model-development uncertainty and
#   are not formal NHANES design-based confidence intervals.
# =====================================================================

import json
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.special import expit
from scipy.optimize import brentq

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    log_loss
)

# ---------------------------------------------------------------------
# 1. REQUIRED OBJECTS
# ---------------------------------------------------------------------

REQUIRED = [
    "OUT",
    "df",
    "dev",
    "test",
    "CLINICAL",
    "available_labs",
    "selected_labs",
    "final_predictions",
    "restricted_predictions",
    "restricted_results",
    "stability_summary",
    "corrected_calibration",
    "evaluate"
]

missing = [x for x in REQUIRED if x not in globals()]

if missing:
    raise RuntimeError(
        "Run V4 and V4.1 first. Missing: " + str(missing)
    )

FINAL_DIR = Path(OUT) / "FINAL_MANUSCRIPT_RESULTS"
FINAL_DIR.mkdir(parents=True, exist_ok=True)

SEED_FINAL = 20261029
RNG_FINAL = np.random.default_rng(SEED_FINAL)
N_BOOT = 2000

def save_csv(data, filename):
    data.to_csv(FINAL_DIR / filename, index=False)

def message(text):
    print(f"[FINAL] {text}", flush=True)

message("FINAL MANUSCRIPT REPORTING STARTED")

# ---------------------------------------------------------------------
# 2. VERIFY TEST POPULATION AND PREDICTIONS
# ---------------------------------------------------------------------

y = test["outcome"].to_numpy(dtype=int)
w = test["weight"].to_numpy(dtype=float)

if not np.isfinite(w).all() or np.any(w <= 0):
    raise ValueError("Invalid temporal survey weights")

if np.unique(y).size != 2:
    raise ValueError("Temporal evaluation lacks both classes")

model_sets = {
    "primary_2005_2016": {
        "clinical": np.asarray(
            final_predictions["clinical"], dtype=float
        ),
        "selected": np.asarray(
            final_predictions["selected"], dtype=float
        ),
        "full": np.asarray(
            final_predictions["full"], dtype=float
        )
    },
    "sensitivity_2007_2016": {
        "clinical": np.asarray(
            restricted_predictions["clinical"], dtype=float
        ),
        "selected": np.asarray(
            restricted_predictions["selected"], dtype=float
        ),
        "full": np.asarray(
            restricted_predictions["full"], dtype=float
        )
    }
}

for analysis_name, predictions in model_sets.items():
    for model_name, p in predictions.items():
        if len(p) != len(test):
            raise ValueError(
                f"Prediction length mismatch: "
                f"{analysis_name}/{model_name}"
            )
        if not np.isfinite(p).all():
            raise ValueError("Nonfinite model predictions")
        if np.any((p < 0) | (p > 1)):
            raise ValueError("Probabilities outside [0,1]")

# ---------------------------------------------------------------------
# 3. POINT ESTIMATES
# ---------------------------------------------------------------------

point_rows = []

for analysis_name, predictions in model_sets.items():
    for model_name, p in predictions.items():
        metrics = evaluate(y, p, w)

        point_rows.append({
            "analysis": analysis_name,
            "model": model_name,
            **metrics
        })

point_estimates = pd.DataFrame(point_rows)

save_csv(
    point_estimates,
    "Table_1_Primary_and_Sensitivity_Performance.csv"
)

message("POINT ESTIMATES")
print(point_estimates.round(5).to_string(index=False))

# ---------------------------------------------------------------------
# 4. STRATIFIED PSU BOOTSTRAP
# ---------------------------------------------------------------------

# Sampling units are resampled WITHIN NHANES design strata.
# The entire participant set of each selected PSU is retained.
#
# This is a fixed-prediction, cluster-bootstrap sensitivity
# analysis. It does NOT refit models in each replicate.
#
# Because the evaluation set contains only the 2017–2018
# cycle, original design-stratum / PSU identifiers are used.

if not {
    "design_stratum",
    "design_psu"
}.issubset(test.columns):
    raise RuntimeError(
        "Survey design variables are missing."
    )

design = test[
    ["design_stratum", "design_psu"]
].copy()

if design.isna().any().any():
    raise RuntimeError("Missing survey design identifiers")

strata_psus = {}

for stratum, part in design.groupby("design_stratum"):
    psu_groups = []

    for psu, subgroup in part.groupby("design_psu"):
        psu_groups.append(
            subgroup.index.to_numpy(dtype=int)
        )

    strata_psus[stratum] = psu_groups

# Confirm the test DataFrame has positional RangeIndex.
if not np.array_equal(
    test.index.to_numpy(),
    np.arange(len(test))
):
    raise RuntimeError(
        "test must have a 0..n-1 index, as produced by V4."
    )

message(
    f"Bootstrap: {N_BOOT} replicates, "
    f"{len(strata_psus)} design strata"
)

def bootstrap_indices(rng):
    positions = []

    for groups in strata_psus.values():
        n_psu = len(groups)

        draws = rng.integers(
            low=0,
            high=n_psu,
            size=n_psu
        )

        for draw in draws:
            positions.extend(groups[draw])

    return np.asarray(positions, dtype=int)

def metric_values(y_sub, p_sub, w_sub):
    if np.unique(y_sub).size != 2:
        return None

    p_sub = np.clip(p_sub, 1e-8, 1 - 1e-8)

    return {
        "auroc": roc_auc_score(
            y_sub, p_sub, sample_weight=w_sub
        ),
        "auprc": average_precision_score(
            y_sub, p_sub, sample_weight=w_sub
        ),
        "brier": brier_score_loss(
            y_sub, p_sub, sample_weight=w_sub
        ),
        "log_loss": log_loss(
            y_sub, p_sub,
            sample_weight=w_sub,
            labels=[0, 1]
        )
    }

COMPARISONS = [
    ("full_vs_clinical", "full", "clinical"),
    ("selected_vs_clinical", "selected", "clinical"),
    ("selected_vs_full", "selected", "full")
]

bootstrap_records = []

for repeat in range(N_BOOT):
    idx = bootstrap_indices(RNG_FINAL)

    y_b = y[idx]
    w_b = w[idx]

    if np.unique(y_b).size != 2:
        continue

    for analysis_name, predictions in model_sets.items():
        metrics_by_model = {}

        for model_name, p in predictions.items():
            metrics_by_model[model_name] = metric_values(
                y_b,
                p[idx],
                w_b
            )

        for comparison, first, second in COMPARISONS:
            for metric in [
                "auroc", "auprc", "brier", "log_loss"
            ]:
                delta = (
                    metrics_by_model[first][metric]
                    - metrics_by_model[second][metric]
                )

                bootstrap_records.append({
                    "replicate": repeat,
                    "analysis": analysis_name,
                    "comparison": comparison,
                    "metric": metric,
                    "delta": float(delta)
                })

    if (repeat + 1) % 500 == 0:
        message(
            f"BOOTSTRAP {repeat + 1}/{N_BOOT}"
        )

bootstrap_df = pd.DataFrame(bootstrap_records)

if bootstrap_df.empty:
    raise RuntimeError("No valid bootstrap replicates")

save_csv(
    bootstrap_df,
    "Supplement_Bootstrap_Replicates.csv"
)

# ---------------------------------------------------------------------
# 5. POINT DIFFERENCES AND PERCENTILE INTERVALS
# ---------------------------------------------------------------------

difference_rows = []

for analysis_name, predictions in model_sets.items():
    metrics_by_model = {
        name: metric_values(y, p, w)
        for name, p in predictions.items()
    }

    for comparison, first, second in COMPARISONS:
        for metric in [
            "auroc", "auprc", "brier", "log_loss"
        ]:
            point_delta = (
                metrics_by_model[first][metric]
                - metrics_by_model[second][metric]
            )

            samples = bootstrap_df.loc[
                (bootstrap_df["analysis"] == analysis_name)
                & (
                    bootstrap_df["comparison"]
                    == comparison
                )
                & (bootstrap_df["metric"] == metric),
                "delta"
            ].to_numpy()

            if len(samples) < 0.90 * N_BOOT:
                raise RuntimeError(
                    "Too few valid bootstrap replicates"
                )

            low, high = np.quantile(
                samples,
                [0.025, 0.975]
            )

            difference_rows.append({
                "analysis": analysis_name,
                "comparison": comparison,
                "metric": metric,
                "point_difference": float(point_delta),
                "bootstrap_lower_95": float(low),
                "bootstrap_upper_95": float(high),
                "valid_replicates": len(samples),
                "interval_method": (
                    "stratified PSU percentile bootstrap; "
                    "fixed predictions"
                )
            })

differences = pd.DataFrame(difference_rows)

save_csv(
    differences,
    "Table_2_Paired_Performance_Differences.csv"
)

message("PAIRED DIFFERENCES AND APPROXIMATE INTERVALS")
print(differences.round(5).to_string(index=False))

# ---------------------------------------------------------------------
# 6. CALIBRATION — SEPARATE DEFINITIONS
# ---------------------------------------------------------------------

def logit(p):
    p = np.clip(
        np.asarray(p, dtype=float),
        1e-7,
        1 - 1e-7
    )
    return np.log(p / (1 - p))

def calibration_in_large(y, p, w):
    lp = logit(p)

    def score(alpha):
        return np.sum(
            w * (y - expit(alpha + lp))
        )

    return float(brentq(score, -40, 40))

# This addendum does not refit or change prediction models.
# The calibration slope below is calculated for reporting only.

from sklearn.linear_model import LogisticRegression

def calibration_slope(y, p, w):
    estimator = LogisticRegression(
        C=1e6,
        max_iter=3000,
        random_state=SEED_FINAL
    )

    estimator.fit(
        logit(p).reshape(-1, 1),
        y,
        sample_weight=w / np.mean(w)
    )

    return float(estimator.coef_[0, 0])

calibration_rows = []

for analysis_name, predictions in model_sets.items():
    for model_name, p in predictions.items():
        calibration_rows.append({
            "analysis": analysis_name,
            "model": model_name,
            "observed_weighted_prevalence": float(
                np.average(y, weights=w)
            ),
            "mean_predicted_risk": float(
                np.average(p, weights=w)
            ),
            "calibration_in_the_large": (
                calibration_in_large(y, p, w)
            ),
            "calibration_slope": (
                calibration_slope(y, p, w)
            )
        })

calibration_final = pd.DataFrame(calibration_rows)

save_csv(
    calibration_final,
    "Table_3_Primary_and_Sensitivity_Calibration.csv"
)

message("CALIBRATION")
print(calibration_final.round(5).to_string(index=False))

# ---------------------------------------------------------------------
# 7. LABORATORY STABILITY — EXISTING RESULTS ONLY
# ---------------------------------------------------------------------

stability_final = stability_summary.copy()

stability_final = stability_final.sort_values(
    "selection_frequency",
    ascending=False
)

save_csv(
    stability_final,
    "Table_4_Laboratory_Selection_Stability.csv"
)

message("SELECTION STABILITY")
print(stability_final.to_string(index=False))

# ---------------------------------------------------------------------
# 8. PRIMARY RESULT FIGURE
# ---------------------------------------------------------------------

primary_auroc = differences.loc[
    (differences["comparison"] == "full_vs_clinical")
    & (differences["metric"] == "auroc")
].copy()

labels = {
    "primary_2005_2016": "Development: 2005–2016",
    "sensitivity_2007_2016": "Development: 2007–2016"
}

primary_auroc["label"] = primary_auroc[
    "analysis"
].map(labels)

primary_auroc = primary_auroc.sort_values(
    "analysis"
).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(8, 4.5))

positions = np.arange(len(primary_auroc))

centers = primary_auroc[
    "point_difference"
].to_numpy()

lower = primary_auroc[
    "bootstrap_lower_95"
].to_numpy()

upper = primary_auroc[
    "bootstrap_upper_95"
].to_numpy()

ax.errorbar(
    centers,
    positions,
    xerr=[
        centers - lower,
        upper - centers
    ],
    fmt="o",
    capsize=5,
    linewidth=2
)

ax.axvline(
    0,
    linestyle="--",
    linewidth=1
)

ax.set_yticks(
    positions,
    primary_auroc["label"]
)

ax.set_xlabel(
    "Incremental AUROC: full laboratory panel minus clinical"
)

ax.set_title(
    "Primary and Sensitivity Analysis"
)

ax.invert_yaxis()

plt.tight_layout()
plt.savefig(
    FINAL_DIR / "Figure_1_Incremental_AUROC.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ---------------------------------------------------------------------
# 9. COHORT SUMMARY AND AUDIT FLAGS
# ---------------------------------------------------------------------

cohort_summary = (
    df.groupby("cycle")
    .agg(
        n=("outcome", "size"),
        cases=("outcome", "sum")
    )
    .reset_index()
)

cohort_summary["unweighted_case_fraction"] = (
    cohort_summary["cases"] /
    cohort_summary["n"]
)

weighted_prevalence = (
    df.groupby("cycle")
    .apply(
        lambda x: np.average(
            x["outcome"],
            weights=x["weight"]
        ),
        include_groups=False
    )
    .reset_index(name="weighted_case_fraction")
)

cohort_summary = cohort_summary.merge(
    weighted_prevalence,
    on="cycle"
)

save_csv(
    cohort_summary,
    "Supplement_Cohort_By_Cycle.csv"
)

# ---------------------------------------------------------------------
# 10. MANUSCRIPT READINESS
# ---------------------------------------------------------------------

readiness = pd.DataFrame([
    {
        "item": "Primary temporal comparison",
        "status": "completed",
        "note": "Full laboratory panel vs clinical"
    },
    {
        "item": "2005–2006 exclusion sensitivity",
        "status": "completed",
        "note": "Rebuilt models from V4.1"
    },
    {
        "item": "Approximate paired PSU bootstrap",
        "status": "completed",
        "note": (
            "Fixed-model uncertainty only; "
            "not formal NHANES design-based inference"
        )
    },
    {
        "item": "100-repeat laboratory stability",
        "status": "completed",
        "note": "Existing V4.1 results"
    },
    {
        "item": "Calibration reporting",
        "status": "completed",
        "note": (
            "Calibration-in-the-large and slope "
            "reported separately"
        )
    },
    {
        "item": "2005–2018 glucose assay comparability",
        "status": "requires_document_review",
        "note": (
            "CDC GLU analytic notes across all cycles"
        )
    },
    {
        "item": "2005–2018 HbA1c assay comparability",
        "status": "requires_document_review",
        "note": (
            "CDC GHB analytic notes across all cycles"
        )
    },
    {
        "item": "Predictor assay comparability",
        "status": "requires_document_review",
        "note": (
            "BIOPRO, lipid and CBC methods"
        )
    },
    {
        "item": "Family history variable meaning",
        "status": "requires_document_review",
        "note": (
            "Verify MCQ300C against cycle codebooks"
        )
    },
    {
        "item": "Survey-design variance estimation",
        "status": "requires_methodological_review",
        "note": (
            "Approximate bootstrap is not a "
            "replacement for full design-based analysis"
        )
    },
    {
        "item": "Independent external validation",
        "status": "not_performed",
        "note": (
            "2017–2018 is an inspected temporal "
            "evaluation, not an untouched external cohort"
        )
    }
])

save_csv(
    readiness,
    "Scientific_Readiness_Checklist.csv"
)

manifest = {
    "analysis": "NHANES Q2 Final Manuscript Reporting",
    "version": "4.2",
    "timestamp_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "seed": SEED_FINAL,
    "bootstrap_replicates_requested": N_BOOT,
    "test_n": int(len(test)),
    "test_cases": int(y.sum()),
    "primary_model": "full",
    "primary_comparator": "clinical",
    "sensitivity_development_period": "2007-2016",
    "temporal_evaluation_period": "2017-2018",
    "selected_labs_original": list(selected_labs),
    "candidate_labs": list(available_labs),
    "new_model_selection_performed": False,
    "new_temporal_model_tuning_performed": False,
    "assay_harmonization_verified": False,
    "formal_survey_design_variance_completed": False,
    "independent_external_validation": False
}

with open(
    FINAL_DIR / "Final_Analysis_Manifest.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        manifest,
        file,
        indent=2,
        ensure_ascii=False
    )

message("=" * 70)
message("FINAL COMPUTATIONAL REPORTING COMPLETE")
message("=" * 70)

print("\nPRIMARY AUROC DIFFERENCES")
print(primary_auroc[
    [
        "label",
        "point_difference",
        "bootstrap_lower_95",
        "bootstrap_upper_95"
    ]
].round(5).to_string(index=False))

print("\nSCIENTIFIC READINESS")
print(readiness.to_string(index=False))

print("\nOUTPUT FOLDER:")
print(FINAL_DIR)

message(
    "No further panel optimization should be performed "
    "on the inspected 2017–2018 evaluation set."
)
```

```python
# ======================================================================
# NHANES 2005–2018
# NON-GLYCEMIC LABORATORY MARKERS FOR UNDIAGNOSED DIABETES
#
# COMPLETE REPRODUCIBLE ANALYSIS — VERSION 4.0
#
# Single-cell Google Colab / standalone Python script
# Author: To be completed by the researcher
#
# DATA SOURCE: US CDC, National Health and Nutrition Examination Survey
# PERIOD: 2005–2006 through 2017–2018
#
# This script downloads all required data from scratch.
# No previous project files, datasets, models, or outputs are required.
#
# PRIMARY OBJECTIVE:
# Assess whether non-glycemic laboratory measurements improve
# prediction of previously undiagnosed diabetes beyond
# routinely available clinical characteristics.
#
# IMPORTANT SCIENTIFIC RESTRICTIONS:
# 1. This is a predictive, observational study.
# 2. No unverified laboratory conversion equations are applied.
# 3. Assay harmonization is explicitly flagged for review.
# 4. Bootstrap intervals are approximate sensitivity intervals,
#    not formal full NHANES design-based confidence intervals.
# 5. The 2017–2018 cycle was previously explored in this project;
#    it must not be described as a fully untouched holdout.
# 6. No actual monetary laboratory costs are assumed.
# 7. No clinical deployment or diagnostic use is established.
# ======================================================================

import os
import sys
import io
import json
import time
import warnings
import subprocess
import importlib
from pathlib import Path
from datetime import datetime, timezone

def require_package(module, package=None):
    try:
        importlib.import_module(module)
    except ImportError:
        subprocess.check_call([
            sys.executable, "-m", "pip", "install", "-q",
            package or module
        ])

require_package("requests")
require_package("sklearn", "scikit-learn")

import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    log_loss,
    roc_curve,
    precision_recall_curve
)

warnings.filterwarnings("ignore", category=FutureWarning)

# ======================================================================
# 01 — CONFIGURATION
# ======================================================================

SEED = 42
RNG = np.random.default_rng(SEED)

CYCLES = {
    "2005-2006": ("D", 2005),
    "2007-2008": ("E", 2007),
    "2009-2010": ("F", 2009),
    "2011-2012": ("G", 2011),
    "2013-2014": ("H", 2013),
    "2015-2016": ("I", 2015),
    "2017-2018": ("J", 2017)
}

TEST_CYCLE = "2017-2018"

OUTER_FOLDS = 4
INNER_FOLDS = 3
BOOTSTRAP_REPEATS = 2000

REGULARIZATION_GRID = [0.01, 0.1, 1.0]

MAX_SELECTED_LABS = 8

# Minimum CV log-loss reduction required to select a new lab.
# This threshold is a prespecified heuristic, not a clinically
# validated minimal important difference.
MIN_LOGLOSS_IMPROVEMENT = 0.0005

MIN_LAB_CYCLE_COVERAGE = 0.75
MIN_CLINICAL_CYCLE_COVERAGE = 0.70

RUN_NESTED_CV = True

# Source documents must be independently verified before changing.
ASSAY_HARMONIZATION_VERIFIED = False
OUTCOME_HARMONIZATION_VERIFIED = True  # glucose bridging and A1c no-adjustment reviewed; other assay review separate
FORMAL_SURVEY_VARIANCE_VERIFIED = False
INDEPENDENT_EXTERNAL_VALIDATION = False

# Always fetch files again. Local cache is not used.
FORCE_REDOWNLOAD = True

# ======================================================================
# 02 — OUTPUT DIRECTORIES
# ======================================================================

try:
    from google.colab import drive
    drive.mount("/content/drive")
    ROOT = Path("/content/drive/MyDrive/Sta")
except ImportError:
    ROOT = Path.cwd()

OUT = ROOT / "NHANES_Q2_FINAL_HARMONIZED"
RAW = OUT / "RAW_CDC_FILES"
TABLES = OUT / "TABLES"
FIGURES = OUT / "FIGURES"
LOGS = OUT / "LOGS"
DATA = OUT / "DATA"

for folder in [OUT, RAW, TABLES, FIGURES, LOGS, DATA]:
    folder.mkdir(parents=True, exist_ok=True)

LOG_MESSAGES = []
DOWNLOAD_RECORDS = []
VARIABLE_RECORDS = []

def log(message):
    line = (
        f"[{datetime.now().strftime('%H:%M:%S')}] "
        f"{message}"
    )
    print(line, flush=True)
    LOG_MESSAGES.append(line)

def save_table(frame, filename):
    frame.to_csv(TABLES / filename, index=False)

def save_plot(filename):
    plt.tight_layout()
    plt.savefig(
        FIGURES / filename,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

log("=" * 76)
log("NHANES Q2 FULL REPRODUCIBLE PIPELINE")
log("=" * 76)
log(f"Output: {OUT}")

# ======================================================================
# 03 — CDC DATA DOWNLOAD
# ======================================================================

session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0 Academic NHANES Research"
})

def get_cdc_file(component, suffix, year, required=False):
    filename = f"{component}_{suffix}.XPT"

    urls = [
        (
            "https://wwwn.cdc.gov/Nchs/Data/Nhanes/"
            f"Public/{year}/DataFiles/{filename}"
        ),
        (
            "https://wwwn.cdc.gov/Nchs/Nhanes/"
            f"{year}-{year+1}/{filename}"
        )
    ]

    errors = []

    for url in urls:
        for attempt in range(3):
            try:
                response = session.get(
                    url,
                    timeout=90
                )
                response.raise_for_status()

                content = response.content

                if len(content) < 100:
                    raise ValueError("Empty or invalid download")

                frame = pd.read_sas(
                    io.BytesIO(content),
                    format="xport"
                )

                if "SEQN" not in [
                    str(x).upper()
                    for x in frame.columns
                ]:
                    raise ValueError("SEQN not found")

                (RAW / filename).write_bytes(content)

                DOWNLOAD_RECORDS.append({
                    "filename": filename,
                    "status": "success",
                    "rows": len(frame),
                    "url": url
                })

                return frame

            except Exception as exc:
                errors.append(str(exc))
                time.sleep(attempt + 1)

    DOWNLOAD_RECORDS.append({
        "filename": filename,
        "status": "failed",
        "rows": 0,
        "url": "; ".join(errors[-3:])
    })

    if required:
        pd.DataFrame(DOWNLOAD_RECORDS).to_csv(
            LOGS / "download_log.csv",
            index=False
        )
        raise RuntimeError(
            f"Required NHANES file unavailable: {filename}\n"
            + "\n".join(errors[-3:])
        )

    log(f"Optional file unavailable: {filename}")
    return None


# ======================================================================
# 04 — NHANES VARIABLE DICTIONARY
# ======================================================================

# Each mapping records actual source variables.
# A missing source variable is never silently fabricated.

VARIABLE_MAP = {
    "DEMO": {
        "age": ["RIDAGEYR"],
        "sex": ["RIAGENDR"],
        "race_ethnicity": ["RIDRETH1"],
        "pregnancy_status": ["RIDEXPRG"],
        "fasting_weight": ["WTSAF2YR"],
        "design_stratum": ["SDMVSTRA"],
        "design_psu": ["SDMVPSU"]
    },
    "DIQ": {
        "diagnosed_diabetes": ["DIQ010"]
    },
    "BMX": {
        "bmi": ["BMXBMI"],
        "waist_circumference": ["BMXWAIST"]
    },
    "GLU": {
        "fasting_glucose": ["LBXGLU"],
        "fasting_weight_fallback": ["WTSAF2YR"]
    },
    "GHB": {
        "hba1c": ["LBXGH"]
    },
    "BIOPRO": {
        "serum_albumin": ["LBXSAL"],
        "serum_creatinine": ["LBXSCR"],
        "uric_acid": ["LBXSUA"],
        "alt": ["LBXSATSI"],
        "ast": ["LBXSASSI"],
        "total_protein": ["LBXSTP"],
        "ggt": ["LBXSGTSI", "LBXSGTS"]
    },
    "CBC": {
        "hemoglobin": ["LBXHGB"],
        "white_blood_cell_count": ["LBXWBCSI"]
    },
    "HDL": {
        "hdl_cholesterol": ["LBDHDD", "LBXHDD"]
    },
    "TCHOL": {
        "total_cholesterol": ["LBXTC"]
    },
    "TRIGLY": {
        "triglycerides": ["LBXTR"]
    },
    "BPX": {
        "sbp1": ["BPXSY1"],
        "sbp2": ["BPXSY2"],
        "sbp3": ["BPXSY3"],
        "sbp4": ["BPXSY4"],
        "dbp1": ["BPXDI1"],
        "dbp2": ["BPXDI2"],
        "dbp3": ["BPXDI3"],
        "dbp4": ["BPXDI4"]
    },
    "SMQ": {
        "smoking_ever": ["SMQ020"]
    },
    "MCQ": {
        "family_history_diabetes": ["MCQ300C"]
    },
    "PAQ": {
        "vigorous_work": ["PAQ605"],
        "moderate_work": ["PAQ620"],
        "vigorous_recreation": ["PAQ650"],
        "moderate_recreation": ["PAQ665"],
        "old_vigorous_activity": ["PAD200"],
        "old_moderate_activity": ["PAD320"]
    }
}

REQUIRED_COMPONENTS = {
    "DEMO", "DIQ", "BMX", "GLU", "GHB"
}

ALL_COMPONENTS = list(VARIABLE_MAP.keys())

def extract_variables(frame, component, cycle):
    if frame is None:
        return None

    frame = frame.copy()
    frame.columns = [
        str(c).upper()
        for c in frame.columns
    ]

    selected = ["SEQN"]
    rename = {}

    for target, candidates in VARIABLE_MAP[component].items():
        matched = next(
            (
                candidate
                for candidate in candidates
                if candidate in frame.columns
            ),
            None
        )

        VARIABLE_RECORDS.append({
            "cycle": cycle,
            "component": component,
            "target_variable": target,
            "source_variable": matched,
            "found": matched is not None
        })

        if matched is not None:
            selected.append(matched)
            rename[matched] = target

    selected = list(dict.fromkeys(selected))

    result = frame[selected].rename(
        columns=rename
    )

    if result["SEQN"].duplicated().any():
        raise ValueError(
            f"Duplicate SEQN in {component}, {cycle}"
        )

    return result


# ======================================================================
# 05 — REBUILD COMPLETE COHORT
# ======================================================================

CORE_CLINICAL = [
    "age",
    "sex",
    "bmi",
    "waist_circumference"
]

OPTIONAL_CLINICAL = [
    "systolic_bp",
    "diastolic_bp",
    "smoking_ever",
    "family_history_diabetes"
]

LABORATORIES = [
    "serum_albumin",
    "serum_creatinine",
    "uric_acid",
    "alt",
    "ast",
    "total_protein",
    "total_cholesterol",
    "hdl_cholesterol",
    "hemoglobin",
    "white_blood_cell_count",
    "triglycerides",
    "ggt"
]

cycle_frames = []
cohort_rows = []

for cycle, (suffix, year) in CYCLES.items():
    log(f"REBUILDING CYCLE: {cycle}")

    components = {}

    for component in ALL_COMPONENTS:
        raw = get_cdc_file(
            component,
            suffix,
            year,
            required=component in REQUIRED_COMPONENTS
        )

        components[component] = extract_variables(
            raw, component, cycle
        )

    frame = components["DEMO"].copy()

    for component in ALL_COMPONENTS:
        if component == "DEMO":
            continue

        piece = components[component]

        if piece is not None:
            frame = frame.merge(
                piece,
                on="SEQN",
                how="left",
                validate="one_to_one"
            )

    frame["cycle"] = cycle

    for column in frame.columns:
        if column != "cycle":
            frame[column] = pd.to_numeric(
                frame[column],
                errors="coerce"
            )

    if "fasting_weight" not in frame.columns:
        frame["fasting_weight"] = np.nan

    if "fasting_weight_fallback" in frame.columns:
        frame["fasting_weight"] = (
            frame["fasting_weight"].fillna(
                frame["fasting_weight_fallback"]
            )
        )

    # Verify mandatory outcome and design variables.
    mandatory = [
        "age", "sex",
        "diagnosed_diabetes",
        "fasting_glucose",
        "hba1c",
        "fasting_weight",
        "design_stratum",
        "design_psu"
    ]

    missing_mandatory = [
        x for x in mandatory
        if x not in frame.columns
    ]

    if missing_mandatory:
        raise RuntimeError(
            f"{cycle}: missing mandatory variables "
            f"{missing_mandatory}"
        )

    for column in (
        CORE_CLINICAL +
        OPTIONAL_CLINICAL +
        LABORATORIES
    ):
        if column not in frame.columns:
            frame[column] = np.nan

    # Blood pressure: mean of available positive readings.
    for prefix, output in [
        ("sbp", "systolic_bp"),
        ("dbp", "diastolic_bp")
    ]:
        cols = [
            f"{prefix}{i}"
            for i in range(1, 5)
            if f"{prefix}{i}" in frame.columns
        ]

        if cols:
            frame[output] = (
                frame[cols]
                .where(frame[cols] > 0)
                .mean(axis=1)
            )

    # Valid questionnaire responses only.
    for column in [
        "smoking_ever",
        "family_history_diabetes"
    ]:
        frame[column] = frame[column].map({
            1: 1.0,
            2: 0.0
        })

    # Physical activity questionnaires changed across cycles.
    # We do not pool the 2005 questions with later questions.
    # Activity is excluded from the primary pooled model.
    # The original responses remain available for audit.

    # Exclusion flags are retained for reproducibility.
    frame["exclude_age"] = ~frame["age"].ge(20)

    # DIQ010 == 2: respondent reports no diabetes diagnosis.
    frame["exclude_diagnosis"] = (
        ~frame["diagnosed_diabetes"].eq(2)
    )

    if "pregnancy_status" in frame.columns:
        frame["exclude_pregnancy"] = (
            frame["pregnancy_status"].eq(1)
        )
    else:
        frame["exclude_pregnancy"] = False

    frame["exclude_missing_outcome"] = (
        frame["hba1c"].isna() |
        frame["fasting_glucose"].isna()
    )

    frame["exclude_weight"] = (
        ~frame["fasting_weight"].gt(0)
    )

    frame["exclude_design"] = (
        frame["design_stratum"].isna() |
        frame["design_psu"].isna()
    )

    exclude_columns = [
        "exclude_age",
        "exclude_diagnosis",
        "exclude_pregnancy",
        "exclude_missing_outcome",
        "exclude_weight",
        "exclude_design"
    ]

    eligible = ~frame[exclude_columns].any(axis=1)

    included = frame.loc[eligible].copy()

    # CDC documented bridging: align all glucose to 2015+ C311 scale.
    # GLU_I already contains C501 specimens adjusted by CDC: NEVER transform 2015-16 again.
    # 2005-06 Hitachi 911 -> 2007-08 ModP -> C311 scale (sensitivity only).
    included["fasting_glucose_original"] = included["fasting_glucose"].copy()
    included["outcome_original"] = (
        included["hba1c"].ge(6.5) |
        included["fasting_glucose_original"].ge(126)
    ).astype(int)
    if cycle == "2005-2006":
        included["fasting_glucose"] = (
            1.023 * (included["fasting_glucose_original"] + 1.148) - 0.5108
        )
        conversion = "2005 Hitachi911 -> ModP (+1.148) -> C311"
    elif cycle in ["2007-2008", "2009-2010", "2011-2012", "2013-2014"]:
        included["fasting_glucose"] = (
            1.023 * included["fasting_glucose_original"] - 0.5108
        )
        conversion = "pre-2015 to C311 (CDC GLU_I bridge)"
    else:
        conversion = "none: already C311 comparable"
    included["glucose_conversion"] = conversion
    included["outcome"] = (
        included["hba1c"].ge(6.5) |
        included["fasting_glucose"].ge(126)
    ).astype(int)
    included["outcome_changed"] = included["outcome"].ne(included["outcome_original"])
    log(f"LABEL AUDIT {cycle}: changed={int(included['outcome_changed'].sum())}, "
        f"0->1={int(((included['outcome_original']==0)&(included['outcome']==1)).sum())}, "
        f"1->0={int(((included['outcome_original']==1)&(included['outcome']==0)).sum())}")

    # Seven two-year cycles: combined 14-year subsample weight.
    included["weight"] = (
        included["fasting_weight"] / 7.0
    )

    included["participant_id"] = (
        cycle + "_" +
        included["SEQN"].astype(int).astype(str)
    )

    included["psu_group"] = (
        cycle + "_" +
        included["design_stratum"].astype(int).astype(str)
        + "_" +
        included["design_psu"].astype(int).astype(str)
    )

    included["stratum_group"] = (
        cycle + "_" +
        included["design_stratum"].astype(int).astype(str)
    )

    cohort_rows.append({
        "cycle": cycle,
        "initial_n": len(frame),
        "excluded_age": int(
            frame["exclude_age"].sum()
        ),
        "excluded_diagnosis": int(
            frame["exclude_diagnosis"].sum()
        ),
        "excluded_pregnancy": int(
            frame["exclude_pregnancy"].sum()
        ),
        "excluded_missing_outcome": int(
            frame["exclude_missing_outcome"].sum()
        ),
        "excluded_invalid_weight": int(
            frame["exclude_weight"].sum()
        ),
        "eligible_n": len(included),
        "cases": int(included["outcome"].sum()),
        "weighted_prevalence": (
            np.average(
                included["outcome"],
                weights=included["weight"]
            )
            if len(included) else np.nan
        ),
        "unique_stratum_psu_pairs": (
            included["psu_group"].nunique()
        )
    })

    if len(included) == 0:
        raise RuntimeError(
            f"No eligible observations in {cycle}"
        )

    cycle_frames.append(included)

df = pd.concat(
    cycle_frames,
    ignore_index=True,
    sort=False
)

if not df["participant_id"].is_unique:
    raise ValueError("Duplicate participant IDs")

if not df["weight"].gt(0).all():
    raise ValueError("Invalid survey weights")

save_table(
    pd.DataFrame(cohort_rows),
    "Table_01_Cohort.csv"
)

save_table(
    pd.DataFrame(VARIABLE_RECORDS),
    "Supplement_Source_Variable_Mapping.csv"
)

pd.DataFrame(DOWNLOAD_RECORDS).to_csv(
    LOGS / "CDC_Download_Log.csv",
    index=False
)

df.to_csv(
    DATA / "NHANES_Rebuilt_Cohort.csv",
    index=False
)

audit_labels = df.groupby("cycle", as_index=False).agg(
    eligible_n=("outcome", "size"),
    cases_before=("outcome_original", "sum"),
    cases_after=("outcome", "sum"),
    labels_changed=("outcome_changed", "sum")
)
audit_labels["changed_0_to_1"] = [
    int(((g.outcome_original == 0) & (g.outcome == 1)).sum())
    for _, g in df.groupby("cycle", sort=True)
]
audit_labels["changed_1_to_0"] = [
    int(((g.outcome_original == 1) & (g.outcome == 0)).sum())
    for _, g in df.groupby("cycle", sort=True)
]
save_table(audit_labels, "Table_00_Glucose_Label_Change_Audit.csv")
print("\nGLUCOSE HARMONIZATION AUDIT")
display(audit_labels)
log(f"TOTAL COHORT N={len(df)}")
log(f"TOTAL CASES={int(df['outcome'].sum())}")

# ======================================================================
# 06 — DEVELOPMENT / TEMPORAL EVALUATION
# ======================================================================

dev = df.loc[
    (~df["cycle"].isin([TEST_CYCLE, "2005-2006"]))
].copy().reset_index(drop=True)

test = df.loc[
    df["cycle"] == TEST_CYCLE
].copy().reset_index(drop=True)

log(f"Development N={len(dev)}")
log(f"Temporal evaluation N={len(test)}")

# ======================================================================
# 07 — FEATURE AVAILABILITY
# ======================================================================

coverage_rows = []

for feature in (
    CORE_CLINICAL +
    OPTIONAL_CLINICAL +
    LABORATORIES +
    ["race_ethnicity"]
):
    for cycle, group in df.groupby("cycle"):
        coverage_rows.append({
            "feature": feature,
            "cycle": cycle,
            "coverage": group[feature].notna().mean()
                if feature in group.columns else 0,
            "median": group[feature].median()
                if feature in group.columns else np.nan
        })

coverage = pd.DataFrame(coverage_rows)

save_table(
    coverage,
    "Table_02_Feature_Coverage.csv"
)

minimum_coverage = (
    coverage.groupby("feature")["coverage"]
    .min()
    .to_dict()
)

missing_core = [
    x for x in CORE_CLINICAL
    if minimum_coverage.get(x, 0) <
    MIN_CLINICAL_CYCLE_COVERAGE
]

if missing_core:
    raise RuntimeError(
        "Insufficient coverage for mandatory predictors: "
        + str(missing_core)
    )

available_optional = [
    x for x in OPTIONAL_CLINICAL
    if minimum_coverage.get(x, 0) >=
    MIN_CLINICAL_CYCLE_COVERAGE
]

available_labs = [
    x for x in LABORATORIES
    if minimum_coverage.get(x, 0) >=
    MIN_LAB_CYCLE_COVERAGE
]

# Preserve the original 12-lab candidate definition.
# Exclusions must be reported explicitly.
excluded_labs = [
    x for x in LABORATORIES
    if x not in available_labs
]

# One-hot encode race/ethnicity, using the same fixed coding
# for development and temporal evaluation.
race_columns = []

if minimum_coverage.get("race_ethnicity", 0) >= 0.70:
    for category in [2, 3, 4, 5]:
        column = f"race_category_{category}"
        race_columns.append(column)

        for frame in [dev, test]:
            frame[column] = np.where(
                frame["race_ethnicity"].notna(),
                (
                    frame["race_ethnicity"] == category
                ).astype(float),
                np.nan
            )

CLINICAL = (
    CORE_CLINICAL +
    available_optional +
    race_columns
)

if len(available_labs) < 3:
    raise RuntimeError(
        "Insufficient laboratory coverage; "
        "review source mapping."
    )

log(f"CLINICAL FEATURES: {CLINICAL}")
log(f"AVAILABLE LABORATORIES: {available_labs}")
log(f"EXCLUDED LABORATORIES: {excluded_labs}")

save_table(
    pd.DataFrame([
        {
            "feature": x,
            "minimum_cycle_coverage":
                minimum_coverage.get(x, np.nan),
            "included_clinical": x in CLINICAL,
            "included_laboratory": x in available_labs
        }
        for x in (
            CORE_CLINICAL +
            OPTIONAL_CLINICAL +
            LABORATORIES
        )
    ]),
    "Table_03_Feature_Eligibility.csv"
)

# ======================================================================
# 08 — MODEL UTILITIES
# ======================================================================

def sample_weights(frame):
    w = frame["weight"].to_numpy(dtype=float)
    return w / np.mean(w)

def model_pipeline(C):
    return Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                keep_empty_features=True
            )
        ),
        ("scaler", StandardScaler()),
        (
            "model",
            LogisticRegression(
                C=C,
                max_iter=3000,
                random_state=SEED
            )
        )
    ])

def fit_model(frame, features, C):
    model = model_pipeline(C)
    model.fit(
        frame[features],
        frame["outcome"],
        model__sample_weight=sample_weights(frame)
    )
    return model

def predict(model, frame, features):
    return model.predict_proba(
        frame[features]
    )[:, 1]

def evaluate(y, p, w):
    y = np.asarray(y, dtype=int)
    p = np.clip(
        np.asarray(p, dtype=float),
        1e-8, 1 - 1e-8
    )
    w = np.asarray(w, dtype=float)

    result = {
        "n": len(y),
        "cases": int(y.sum()),
        "weighted_prevalence": float(
            np.average(y, weights=w)
        ),
        "mean_predicted_risk": float(
            np.average(p, weights=w)
        ),
        "brier": float(
            brier_score_loss(
                y, p, sample_weight=w
            )
        ),
        "log_loss": float(
            log_loss(
                y, p,
                sample_weight=w,
                labels=[0, 1]
            )
        ),
        "auroc": np.nan,
        "auprc": np.nan
    }

    if np.unique(y).size == 2:
        result["auroc"] = float(
            roc_auc_score(
                y, p, sample_weight=w
            )
        )
        result["auprc"] = float(
            average_precision_score(
                y, p, sample_weight=w
            )
        )

    return result

def folds_for(frame, k, seed):
    splitter = StratifiedGroupKFold(
        n_splits=k,
        shuffle=True,
        random_state=seed
    )

    return list(
        splitter.split(
            frame,
            frame["outcome"],
            frame["psu_group"]
        )
    )

def cv_losses(frame, features, C, folds):
    losses = []

    for train_idx, valid_idx in folds:
        train = frame.iloc[train_idx]
        valid = frame.iloc[valid_idx]

        if (
            train["outcome"].nunique() < 2 or
            valid["outcome"].nunique() < 2
        ):
            return np.array([np.inf])

        model = fit_model(
            train, features, C
        )

        p = predict(
            model, valid, features
        )

        losses.append(
            log_loss(
                valid["outcome"],
                p,
                sample_weight=valid["weight"],
                labels=[0, 1]
            )
        )

    return np.asarray(losses)

def tune(frame, features, folds):
    results = []

    for C in REGULARIZATION_GRID:
        losses = cv_losses(
            frame, features, C, folds
        )

        results.append({
            "C": C,
            "mean_loss": float(np.mean(losses)),
            "fold_losses": losses
        })

    return min(
        results,
        key=lambda x: x["mean_loss"]
    )

# ======================================================================
# 09 — SEQUENTIAL LABORATORY SELECTION
# ======================================================================

def select_laboratories(
    frame, clinical, candidates, seed
):
    folds = folds_for(
        frame,
        INNER_FOLDS,
        seed
    )

    baseline = tune(
        frame, clinical, folds
    )

    selected = []
    history = []

    current_loss = baseline["mean_loss"]

    for step in range(MAX_SELECTED_LABS):
        remaining = [
            x for x in candidates
            if x not in selected
        ]

        if not remaining:
            break

        options = []

        for lab in remaining:
            features = (
                clinical + selected + [lab]
            )

            candidate = tune(
                frame, features, folds
            )

            improvement = (
                current_loss -
                candidate["mean_loss"]
            )

            options.append({
                "laboratory": lab,
                "mean_loss": candidate["mean_loss"],
                "improvement": improvement,
                "C": candidate["C"]
            })

        best = max(
            options,
            key=lambda x: x["improvement"]
        )

        if (
            best["improvement"] <
            MIN_LOGLOSS_IMPROVEMENT
        ):
            break

        selected.append(
            best["laboratory"]
        )

        current_loss = best["mean_loss"]

        history.append({
            "step": step + 1,
            **best
        })

    final = tune(
        frame,
        clinical + selected,
        folds
    )

    return {
        "selected": selected,
        "history": history,
        "C": final["C"],
        "cv_loss": final["mean_loss"]
    }

# ======================================================================
# 10 — GROUP-AWARE NESTED CROSS-VALIDATION
# ======================================================================

nested_rows = []
selection_rows = []

if RUN_NESTED_CV:
    log("STARTING NESTED CROSS-VALIDATION")

    outer_folds = folds_for(
        dev, OUTER_FOLDS, SEED
    )

    for fold_number, (tr, va) in enumerate(
        outer_folds, 1
    ):
        log(
            f"OUTER FOLD {fold_number}/{OUTER_FOLDS}"
        )

        outer_train = dev.iloc[tr].copy()
        outer_valid = dev.iloc[va].copy()

        selection = select_laboratories(
            outer_train,
            CLINICAL,
            available_labs,
            SEED + fold_number
        )

        for lab in selection["selected"]:
            selection_rows.append({
                "fold": fold_number,
                "laboratory": lab
            })

        inner = folds_for(
            outer_train,
            INNER_FOLDS,
            SEED + 100 + fold_number
        )

        definitions = {
            "clinical": CLINICAL,
            "selected": (
                CLINICAL + selection["selected"]
            ),
            "full": (
                CLINICAL + available_labs
            )
        }

        for name, features in definitions.items():
            tuned = tune(
                outer_train,
                features,
                inner
            )

            model = fit_model(
                outer_train,
                features,
                tuned["C"]
            )

            p = predict(
                model,
                outer_valid,
                features
            )

            nested_rows.append({
                "fold": fold_number,
                "model": name,
                "C": tuned["C"],
                "selected_labs": ",".join(
                    selection["selected"]
                ),
                **evaluate(
                    outer_valid["outcome"],
                    p,
                    outer_valid["weight"]
                )
            })

    nested = pd.DataFrame(nested_rows)

    save_table(
        nested,
        "Table_04_Nested_CV.csv"
    )

    frequency = (
        pd.DataFrame(selection_rows)
        .groupby("laboratory")
        .size()
        .reset_index(name="selected_folds")
    )

    frequency["frequency"] = (
        frequency["selected_folds"] /
        OUTER_FOLDS
    )

    save_table(
        frequency,
        "Table_05_Selection_Stability.csv"
    )

    log("NESTED CV COMPLETE")
    display(nested.round(5))
    display(frequency)

# ======================================================================
# 11 — INDEPENDENT CALIBRATION PARTITION
# ======================================================================

log("CREATING GROUP-SEPARATED CALIBRATION DATA")

# A group-aware split is made before final variable selection.
# No calibration participant contributes to selection or tuning.

splitter = StratifiedGroupKFold(
    n_splits=4,
    shuffle=True,
    random_state=SEED + 500
)

train_idx, cal_idx = next(
    splitter.split(
        dev,
        dev["outcome"],
        dev["psu_group"]
    )
)

train = dev.iloc[
    train_idx
].copy().reset_index(drop=True)

cal = dev.iloc[
    cal_idx
].copy().reset_index(drop=True)

assert set(train["psu_group"]).isdisjoint(
    set(cal["psu_group"])
)

if (
    train["outcome"].nunique() < 2 or
    cal["outcome"].nunique() < 2
):
    raise RuntimeError(
        "Training/calibration split lacks outcome classes."
    )

log(f"TRAIN N={len(train)}")
log(f"CALIBRATION N={len(cal)}")

# ======================================================================
# 12 — FINAL LABORATORY SELECTION
# ======================================================================

selection = select_laboratories(
    train,
    CLINICAL,
    available_labs,
    SEED + 700
)

selected_labs = selection["selected"]

save_table(
    pd.DataFrame(selection["history"]),
    "Table_06_Final_Laboratory_Selection.csv"
)

log(f"SELECTED LABORATORIES: {selected_labs}")

# ======================================================================
# 13 — MODEL TRAINING AND CALIBRATION
# ======================================================================

definitions = {
    "clinical": CLINICAL,
    "selected": (
        CLINICAL + selected_labs
    ),
    "full": (
        CLINICAL + available_labs
    )
}

final_predictions = {}
final_regularization = {}

inner = folds_for(
    train,
    INNER_FOLDS,
    SEED + 900
)

def logit_transform(p):
    p = np.clip(
        np.asarray(p),
        1e-6, 1 - 1e-6
    )
    return np.log(
        p / (1 - p)
    ).reshape(-1, 1)

for name, features in definitions.items():
    log(f"FITTING MODEL: {name}")

    tuned = tune(
        train,
        features,
        inner
    )

    model = fit_model(
        train,
        features,
        tuned["C"]
    )

    p_cal = predict(
        model, cal, features
    )

    p_test = predict(
        model, test, features
    )

    recalibrator = LogisticRegression(
        C=1e6,
        max_iter=3000,
        random_state=SEED
    )

    recalibrator.fit(
        logit_transform(p_cal),
        cal["outcome"],
        sample_weight=sample_weights(cal)
    )

    p_final = recalibrator.predict_proba(
        logit_transform(p_test)
    )[:, 1]

    final_predictions[name] = p_final
    final_regularization[name] = tuned["C"]

# ======================================================================
# 14 — TEMPORAL PERFORMANCE
# ======================================================================

y = test["outcome"].to_numpy(dtype=int)
w = test["weight"].to_numpy(dtype=float)

performance = pd.DataFrame([
    {
        "model": name,
        "C": final_regularization[name],
        "laboratory_count": (
            len(definitions[name]) -
            len(CLINICAL)
        ),
        **evaluate(y, p, w)
    }
    for name, p in final_predictions.items()
])

save_table(
    performance,
    "Table_07_Temporal_Performance.csv"
)

log("TEMPORAL PERFORMANCE")
display(performance.round(5))

# ======================================================================
# 15 — PAIRED COMPARISONS
# ======================================================================

comparisons = [
    ("selected_vs_clinical", "selected", "clinical"),
    ("full_vs_clinical", "full", "clinical"),
    ("selected_vs_full", "selected", "full")
]

comparison_rows = []

for label, a, b in comparisons:
    ma = evaluate(y, final_predictions[a], w)
    mb = evaluate(y, final_predictions[b], w)

    comparison_rows.append({
        "comparison": label,
        **{
            f"delta_{metric}": (
                ma[metric] - mb[metric]
            )
            for metric in [
                "auroc", "auprc",
                "brier", "log_loss"
            ]
        }
    })

comparison_table = pd.DataFrame(
    comparison_rows
)

save_table(
    comparison_table,
    "Table_08_Model_Differences.csv"
)

display(comparison_table.round(5))

# ======================================================================
# 16 — STRATIFIED PSU BOOTSTRAP
# ======================================================================

log("RUNNING APPROXIMATE PSU BOOTSTRAP")

strata = test["stratum_group"].to_numpy()
psus = test["psu_group"].to_numpy()

clusters = {}

for stratum in np.unique(strata):
    positions = np.where(
        strata == stratum
    )[0]

    clusters[stratum] = [
        positions[psus[positions] == psu]
        for psu in np.unique(psus[positions])
    ]

bootstrap_rows = []

for repeat in range(BOOTSTRAP_REPEATS):
    sampled = []

    for stratum, units in clusters.items():
        choices = RNG.integers(
            0,
            len(units),
            size=len(units)
        )

        for choice in choices:
            sampled.extend(
                units[choice].tolist()
            )

    idx = np.asarray(
        sampled, dtype=int
    )

    if np.unique(y[idx]).size < 2:
        continue

    replicate_metrics = {
        name: evaluate(
            y[idx],
            p[idx],
            w[idx]
        )
        for name, p in final_predictions.items()
    }

    for label, a, b in comparisons:
        for metric in [
            "auroc", "auprc",
            "brier", "log_loss"
        ]:
            bootstrap_rows.append({
                "repeat": repeat,
                "comparison": label,
                "metric": metric,
                "difference": (
                    replicate_metrics[a][metric] -
                    replicate_metrics[b][metric]
                )
            })

bootstrap = pd.DataFrame(
    bootstrap_rows
)

bootstrap.to_csv(
    DATA / "Bootstrap_Replicates.csv",
    index=False
)

intervals = (
    bootstrap.groupby(
        ["comparison", "metric"]
    )["difference"]
    .agg(
        bootstrap_mean="mean",
        lower_95=lambda x: x.quantile(0.025),
        upper_95=lambda x: x.quantile(0.975),
        valid_replicates="count"
    )
    .reset_index()
)

save_table(
    intervals,
    "Table_09_Approximate_Bootstrap_Intervals.csv"
)

display(intervals.round(5))

# ======================================================================
# 17 — CALIBRATION
# ======================================================================

calibration_rows = []
calibration_bins = []

for name, p in final_predictions.items():
    recalibration_model = LogisticRegression(
        C=1e6,
        max_iter=3000,
        random_state=SEED
    )

    recalibration_model.fit(
        logit_transform(p),
        y,
        sample_weight=w / np.mean(w)
    )

    calibration_rows.append({
        "model": name,
        "observed_prevalence": np.average(
            y, weights=w
        ),
        "mean_predicted_risk": np.average(
            p, weights=w
        ),
        "calibration_intercept": (
            recalibration_model.intercept_[0]
        ),
        "calibration_slope": (
            recalibration_model.coef_[0, 0]
        )
    })

    bins = pd.qcut(
        pd.Series(p).rank(method="first"),
        10,
        labels=False,
        duplicates="drop"
    ).to_numpy()

    for bin_id in np.unique(bins):
        mask = bins == bin_id

        calibration_bins.append({
            "model": name,
            "bin": int(bin_id) + 1,
            "n": int(mask.sum()),
            "predicted": np.average(
                p[mask],
                weights=w[mask]
            ),
            "observed": np.average(
                y[mask],
                weights=w[mask]
            )
        })

calibration = pd.DataFrame(
    calibration_rows
)

calibration_curve = pd.DataFrame(
    calibration_bins
)

save_table(
    calibration,
    "Table_10_Calibration.csv"
)

save_table(
    calibration_curve,
    "Supplement_Calibration_Bins.csv"
)

display(calibration.round(5))

# ======================================================================
# 18 — SUBGROUPS
# ======================================================================

test["age_group"] = pd.cut(
    test["age"],
    bins=[19, 39, 59, np.inf],
    labels=["20-39", "40-59", "60+"]
)

test["sex_group"] = test["sex"].map({
    1: "Male",
    2: "Female"
})

subgroup_rows = []

for variable in ["age_group", "sex_group"]:
    values = test[variable].astype(str).to_numpy()

    for group in sorted(np.unique(values)):
        if group == "nan":
            continue

        mask = values == group

        for name, p in final_predictions.items():
            subgroup_rows.append({
                "variable": variable,
                "group": group,
                "model": name,
                **evaluate(
                    y[mask],
                    p[mask],
                    w[mask]
                )
            })

subgroups = pd.DataFrame(
    subgroup_rows
)

save_table(
    subgroups,
    "Table_11_Subgroups.csv"
)

display(subgroups.round(5))

# ======================================================================
# 19 — COMMON COMPLETE-CASE SENSITIVITY
# ======================================================================

all_model_features = list(
    dict.fromkeys(
        CLINICAL + available_labs
    )
)

complete_mask = (
    test[all_model_features]
    .notna()
    .all(axis=1)
    .to_numpy()
)

complete_rows = []

for name, p in final_predictions.items():
    complete_rows.append({
        "model": name,
        "complete_case_fraction": (
            complete_mask.mean()
        ),
        **evaluate(
            y[complete_mask],
            p[complete_mask],
            w[complete_mask]
        )
    })

save_table(
    pd.DataFrame(complete_rows),
    "Table_12_Common_Complete_Case.csv"
)

# ======================================================================
# 20 — DECISION CURVE ANALYSIS
# ======================================================================

thresholds = np.linspace(
    0.01, 0.15, 29
)

weighted_prevalence = np.average(
    y, weights=w
)

wn = w / w.sum()

dca_rows = []

for threshold in thresholds:
    dca_rows.append({
        "threshold": threshold,
        "strategy": "treat_none",
        "net_benefit": 0.0
    })

    dca_rows.append({
        "threshold": threshold,
        "strategy": "treat_all",
        "net_benefit": (
            weighted_prevalence -
            (1 - weighted_prevalence) *
            threshold / (1 - threshold)
        )
    })

    for name, p in final_predictions.items():
        predicted_positive = p >= threshold

        tp = np.sum(
            wn * (
                (y == 1) &
                predicted_positive
            )
        )

        fp = np.sum(
            wn * (
                (y == 0) &
                predicted_positive
            )
        )

        dca_rows.append({
            "threshold": threshold,
            "strategy": name,
            "net_benefit": (
                tp -
                fp * threshold / (1 - threshold)
            )
        })

dca = pd.DataFrame(dca_rows)

save_table(
    dca,
    "Table_13_Decision_Curve.csv"
)

# ======================================================================
# 21 — PUBLICATION FIGURES
# ======================================================================

log("CREATING PUBLICATION FIGURES")

plt.figure(figsize=(7, 6))

for name, p in final_predictions.items():
    fpr, tpr, _ = roc_curve(
        y, p, sample_weight=w
    )

    auc = roc_auc_score(
        y, p, sample_weight=w
    )

    plt.plot(
        fpr, tpr,
        linewidth=2,
        label=f"{name}: AUROC {auc:.3f}"
    )

plt.plot(
    [0, 1], [0, 1],
    linestyle="--",
    linewidth=1
)

plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("Temporal Validation: ROC")
plt.legend()
save_plot("Figure_01_ROC.png")

plt.figure(figsize=(7, 6))

for name, p in final_predictions.items():
    precision, recall, _ = precision_recall_curve(
        y, p, sample_weight=w
    )

    ap = average_precision_score(
        y, p, sample_weight=w
    )

    plt.plot(
        recall, precision,
        linewidth=2,
        label=f"{name}: AUPRC {ap:.3f}"
    )

plt.axhline(
    weighted_prevalence,
    linestyle="--",
    linewidth=1
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Temporal Validation: Precision-Recall")
plt.legend()
save_plot("Figure_02_Precision_Recall.png")

plt.figure(figsize=(7, 6))

for name, group in calibration_curve.groupby("model"):
    plt.plot(
        group["predicted"],
        group["observed"],
        marker="o",
        linewidth=2,
        label=name
    )

plt.plot(
    [0, 0.35],
    [0, 0.35],
    linestyle="--",
    linewidth=1
)

plt.xlabel("Predicted risk")
plt.ylabel("Observed risk")
plt.title("Temporal Calibration")
plt.legend()
save_plot("Figure_03_Calibration.png")

plt.figure(figsize=(8, 5))

for name, group in dca.groupby("strategy"):
    plt.plot(
        group["threshold"],
        group["net_benefit"],
        linewidth=2,
        label=name
    )

plt.axhline(
    0, linestyle=":", linewidth=1
)

plt.xlabel("Risk threshold")
plt.ylabel("Weighted net benefit")
plt.title("Exploratory Decision Curve Analysis")
plt.legend(fontsize=8)
save_plot("Figure_04_Decision_Curve.png")

# ======================================================================
# 22 — PREDICTIONS AND REPRODUCIBILITY
# ======================================================================

prediction_output = test[
    [
        "participant_id",
        "cycle",
        "SEQN",
        "outcome",
        "weight",
        "design_stratum",
        "design_psu"
    ]
].copy()

for name, p in final_predictions.items():
    prediction_output[
        f"probability_{name}"
    ] = p

prediction_output.to_csv(
    DATA / "Temporal_Predictions.csv",
    index=False
)

# ======================================================================
# 23 — SCIENTIFIC QUALITY GATES
# ======================================================================

quality = pd.DataFrame([
    {
        "check": "All seven cycles available",
        "passed": df["cycle"].nunique() == 7
    },
    {
        "check": "Unique participants",
        "passed": df["participant_id"].is_unique
    },
    {
        "check": "Positive fasting weights",
        "passed": df["weight"].gt(0).all()
    },
    {
        "check": "Outcome markers excluded from predictors",
        "passed": not any(
            x in (
                CLINICAL + available_labs
            )
            for x in [
                "hba1c", "fasting_glucose"
            ]
        )
    },
    {
        "check": "Calibration separated before selection",
        "passed": set(
            train["participant_id"]
        ).isdisjoint(
            set(cal["participant_id"])
        )
    },
    {
        "check": "PSU-group separation",
        "passed": set(
            train["psu_group"]
        ).isdisjoint(
            set(cal["psu_group"])
        )
    },
    {
        "check": "Nested CV completed",
        "passed": RUN_NESTED_CV
    },
    {
        "check": "Calibration reported",
        "passed": len(calibration) == 3
    },
    {
        "check": "Outcome assay harmonization verified",
        "passed": OUTCOME_HARMONIZATION_VERIFIED
    },
    {
        "check": "Predictor assay harmonization verified",
        "passed": ASSAY_HARMONIZATION_VERIFIED
    },
    {
        "check": "Formal survey-design variance completed",
        "passed": FORMAL_SURVEY_VARIANCE_VERIFIED
    },
    {
        "check": "Independent external validation",
        "passed": INDEPENDENT_EXTERNAL_VALIDATION
    },
    {
        "check": "Temporal evaluation fully untouched",
        "passed": False
    }
])

save_table(
    quality,
    "Scientific_Quality_Gates.csv"
)

# ======================================================================
# 24 — SAVE ANALYSIS MANIFEST
# ======================================================================

manifest = {
    "study": (
        "Incremental Predictive Value of Non-Glycemic "
        "Laboratory Markers for Undiagnosed Diabetes"
    ),
    "created_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "seed": SEED,
    "cycles": list(CYCLES.keys()),
    "development_cycles": [
        x for x in CYCLES
        if x not in (TEST_CYCLE, "2005-2006")
    ],
    "temporal_cycle": TEST_CYCLE,
    "total_n": len(df),
    "total_cases": int(df["outcome"].sum()),
    "development_n": len(dev),
    "temporal_n": len(test),
    "temporal_cases": int(test["outcome"].sum()),
    "clinical_features": CLINICAL,
    "candidate_laboratories": LABORATORIES,
    "eligible_laboratories": available_labs,
    "excluded_laboratories": excluded_labs,
    "selected_laboratories": selected_labs,
    "regularization": final_regularization,
    "selection_minimum_improvement":
        MIN_LOGLOSS_IMPROVEMENT,
    "nested_outer_folds": OUTER_FOLDS,
    "nested_inner_folds": INNER_FOLDS,
    "bootstrap_repeats": BOOTSTRAP_REPEATS,
    "survey_inference": (
        "Approximate fixed-prediction PSU bootstrap; "
        "not formal full design-based variance"
    ),
    "assay_harmonization_verified":
        ASSAY_HARMONIZATION_VERIFIED,
    "outcome_harmonization_verified":
        OUTCOME_HARMONIZATION_VERIFIED,
    "glucose_scale": "C311; CDC GLU_I forward, GLU_E bridge for 2005",
    "2005_2006_role": "sensitivity_only",
    "external_validation":
        INDEPENDENT_EXTERNAL_VALIDATION
}

with open(
    OUT / "Reproducibility_Manifest.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        manifest,
        f,
        indent=2,
        ensure_ascii=False
    )

# ======================================================================
# 25 — SOURCE DOCUMENTATION LINKS
# ======================================================================

documentation_rows = []

for cycle, (suffix, year) in CYCLES.items():
    for component in ALL_COMPONENTS:
        documentation_rows.append({
            "cycle": cycle,
            "component": component,
            "documentation_url": (
                "https://wwwn.cdc.gov/Nchs/Data/Nhanes/"
                f"Public/{year}/DataFiles/"
                f"{component}_{suffix}.htm"
            ),
            "analytic_notes_reviewed": False,
            "assay_comparability_verified": False,
            "adjustment_applied": False
        })

save_table(
    pd.DataFrame(documentation_rows),
    "Supplement_CDC_Documentation_Audit.csv"
)

# ======================================================================
# 26 — FINAL RESULTS AND LOGS
# ======================================================================

log("=" * 76)
log("FINAL ANALYSIS RESULTS")
log("=" * 76)

print("\nCOHORT")
display(pd.DataFrame(cohort_rows).round(5))

print("\nFEATURES")
display(
    pd.read_csv(
        TABLES / "Table_03_Feature_Eligibility.csv"
    )
)

print("\nSELECTED LABORATORIES")
print(selected_labs)

print("\nTEMPORAL PERFORMANCE")
display(performance.round(5))

print("\nMODEL DIFFERENCES")
display(comparison_table.round(5))

print("\nBOOTSTRAP INTERVALS")
display(intervals.round(5))

print("\nCALIBRATION")
display(calibration.round(5))

print("\nSUBGROUPS")
display(subgroups.round(5))

print("\nSCIENTIFIC QUALITY GATES")
display(quality)

log(f"ALL RESULTS SAVED TO: {OUT}")

if not quality["passed"].all():
    log(
        "COMPUTATIONAL ANALYSIS COMPLETED. "
        "SCIENTIFIC SUBMISSION REQUIREMENTS REMAIN."
    )

log("END OF REPRODUCIBLE PIPELINE")

(LOGS / "Execution_Log.txt").write_text(
    "\n".join(LOG_MESSAGES),
    encoding="utf-8"
)


# ======================================================================
# 27 — HISTORICAL DEVELOPMENT SENSITIVITY + FINAL PAIRED REPORTING
# ======================================================================
# Primary above: 2007-2016 development. Sensitivity below: 2005-2016.
# Both use the SAME inspected 2017-2018 temporal evaluation.
# No additional panel search on the temporal evaluation set.

from scipy.special import expit
from scipy.optimize import brentq

FINAL_DIR = OUT / "FINAL_MANUSCRIPT_RESULTS"
FINAL_DIR.mkdir(parents=True, exist_ok=True)

def final_fit_predict(training, calibration, evaluation, features, seed):
    folds = folds_for(training, INNER_FOLDS, seed)
    best = tune(training, features, folds)
    mdl = fit_model(training, features, best["C"])
    p_cal = predict(mdl, calibration, features)
    p_eval = predict(mdl, evaluation, features)
    calibrator = LogisticRegression(C=1e6, max_iter=3000, random_state=seed)
    calibrator.fit(logit_transform(p_cal), calibration["outcome"],
                   sample_weight=sample_weights(calibration))
    return calibrator.predict_proba(logit_transform(p_eval))[:, 1]

historical_dev = df.loc[df["cycle"] != TEST_CYCLE].copy().reset_index(drop=True)
for category in [2, 3, 4, 5]:
    col = f"race_category_{category}"
    if col in CLINICAL:
        historical_dev[col] = np.where(
            historical_dev["race_ethnicity"].notna(),
            historical_dev["race_ethnicity"].eq(category).astype(float), np.nan)

historical_split = StratifiedGroupKFold(n_splits=4, shuffle=True, random_state=20261030)
h_tr, h_cal = next(historical_split.split(
    historical_dev, historical_dev["outcome"], historical_dev["psu_group"]))
historical_train = historical_dev.iloc[h_tr].copy().reset_index(drop=True)
historical_cal = historical_dev.iloc[h_cal].copy().reset_index(drop=True)
assert set(historical_train["psu_group"]).isdisjoint(set(historical_cal["psu_group"]))
assert historical_train["outcome"].nunique() == historical_cal["outcome"].nunique() == 2

historical_predictions = {}
for model_name, cols in {
    "clinical": CLINICAL,
    "full": CLINICAL + available_labs
}.items():
    log("HISTORICAL SENSITIVITY FIT: " + model_name)
    historical_predictions[model_name] = final_fit_predict(
        historical_train, historical_cal, test, cols, 20261030)

model_sets = {
    "primary_2007_2016": {
        "clinical": np.asarray(final_predictions["clinical"]),
        "full": np.asarray(final_predictions["full"])
    },
    "sensitivity_2005_2016": historical_predictions
}

def final_metrics(yy, pp, ww):
    return {
        "auroc": float(roc_auc_score(yy, pp, sample_weight=ww)),
        "auprc": float(average_precision_score(yy, pp, sample_weight=ww)),
        "brier": float(brier_score_loss(yy, pp, sample_weight=ww)),
        "log_loss": float(log_loss(yy, np.clip(pp,1e-8,1-1e-8),
                                   sample_weight=ww, labels=[0,1]))
    }

def final_calibration(yy, pp, ww):
    z = logit_transform(pp).ravel()
    alpha = brentq(lambda a: np.sum(ww * (yy - expit(a + z))), -40, 40)
    reg = LogisticRegression(C=1e6, max_iter=3000, random_state=SEED)
    reg.fit(z.reshape(-1,1), yy, sample_weight=ww / ww.mean())
    return {"calibration_in_large": float(alpha),
            "calibration_slope": float(reg.coef_[0,0]),
            "observed_prevalence": float(np.average(yy, weights=ww)),
            "mean_predicted_risk": float(np.average(pp, weights=ww))}

final_point_rows = []
for period, models in model_sets.items():
    for name, prob in models.items():
        final_point_rows.append({"analysis": period, "model": name,
            **final_metrics(y, prob, w), **final_calibration(y, prob, w)})
final_points = pd.DataFrame(final_point_rows)
final_points.to_csv(FINAL_DIR / "Table_1_Performance_And_Calibration.csv", index=False)

# Paired fixed-prediction PSU bootstrap within 2017-2018 strata.
# Not a formal NHANES replicate-weight design-based confidence interval.
psu_groups = {}
for stratum, subset in test.groupby("design_stratum"):
    psu_groups[stratum] = [grp.index.to_numpy(dtype=int)
                           for _, grp in subset.groupby("design_psu")]
assert np.array_equal(test.index.to_numpy(), np.arange(len(test)))
boot_rng = np.random.default_rng(20261030)
boot_rows = []
for b in range(2000):
    idx = np.concatenate([
        np.concatenate([groups[j] for j in boot_rng.integers(0,len(groups),size=len(groups))])
        for groups in psu_groups.values()
    ])
    if np.unique(y[idx]).size != 2:
        continue
    for period, models in model_sets.items():
        a = final_metrics(y[idx], models["full"][idx], w[idx])
        c = final_metrics(y[idx], models["clinical"][idx], w[idx])
        for metric in ("auroc", "auprc", "brier", "log_loss"):
            boot_rows.append({"replicate":b, "analysis":period,
                              "comparison":"full_minus_clinical",
                              "metric":metric, "difference":a[metric]-c[metric]})
    if (b+1)%500==0:
        log(f"FINAL BOOTSTRAP {b+1}/2000")
boot_final = pd.DataFrame(boot_rows)
boot_final.to_csv(FINAL_DIR / "Supplement_Paired_PSU_Bootstrap_Replicates.csv", index=False)
interval_rows=[]
for period, models in model_sets.items():
    a=final_metrics(y,models["full"],w)
    c=final_metrics(y,models["clinical"],w)
    for metric in ("auroc","auprc","brier","log_loss"):
        vals=boot_final.loc[(boot_final.analysis==period)&(boot_final.metric==metric),"difference"]
        if len(vals)<1800: raise RuntimeError("Too few valid bootstrap replicates")
        lo,hi=np.quantile(vals,[0.025,0.975])
        interval_rows.append({"analysis":period,"metric":metric,
                              "point_difference":a[metric]-c[metric],
                              "lower_95":lo,"upper_95":hi,"valid_replicates":len(vals),
                              "method":"fixed-prediction stratified PSU bootstrap (approximate)"})
final_intervals=pd.DataFrame(interval_rows)
final_intervals.to_csv(FINAL_DIR/"Table_2_Full_Minus_Clinical_Intervals.csv",index=False)

fig,ax=plt.subplots(figsize=(8,4))
sub=final_intervals.loc[final_intervals.metric=="auroc"].reset_index(drop=True)
for j,row in sub.iterrows():
    ax.errorbar(row.point_difference,j,xerr=[[row.point_difference-row.lower_95],
               [row.upper_95-row.point_difference]],fmt="o",capsize=4)
ax.set_yticks(range(len(sub)),sub.analysis)
ax.axvline(0,linestyle="--",color="gray")
ax.set_xlabel("Full laboratory minus clinical AUROC")
plt.tight_layout()
fig.savefig(FINAL_DIR/"Figure_1_Incremental_AUROC.png",dpi=300,bbox_inches="tight")
plt.show()
plt.close(fig)

final_manifest={
    "primary_development":"2007-2016", "sensitivity_development":"2005-2016",
    "temporal_evaluation":"2017-2018 (previously inspected)",
    "glucose_harmonization":"CDC C311 scale; 2015-2018 not transformed again",
    "glucose_labels_changed_total":int(df.outcome_changed.sum()),
    "hba1c":"original values; no 2005-06 crossover correction",
    "temporal_test_n":int(len(test)),"temporal_test_cases":int(y.sum()),
    "candidate_labs":available_labs,"primary_models":["clinical","full"],
    "bootstrap_replicates_requested":2000,
    "bootstrap_interpretation":"approximate fixed-prediction cluster intervals",
    "predictor_assay_comparability_fully_verified":False,
    "formal_nhanes_design_based_inference":False,
    "independent_external_validation":False,
    "completed_utc":datetime.now(timezone.utc).isoformat(),
    "cdc_sources":[
        "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2005/DataFiles/GLU_D.htm",
        "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2007/DataFiles/GLU_E.htm",
        "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2015/DataFiles/GLU_I.htm",
        "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2017/DataFiles/GLU_J.htm",
        "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2007/DataFiles/GHB_E.htm"
    ]
}
(FINAL_DIR/"Final_Analysis_Manifest.json").write_text(
    json.dumps(final_manifest,indent=2,ensure_ascii=False),encoding="utf-8")
print("\nFINAL GLUCOSE LABEL AUDIT")
display(audit_labels)
print("\nFINAL PERFORMANCE AND CALIBRATION")
display(final_points.round(5))
print("\nFINAL PAIRED INTERVALS")
display(final_intervals.round(5))
log("FINAL COMPUTATIONAL ANALYSIS COMPLETE: " + str(FINAL_DIR))
```

```python
# =====================================================================
# NHANES Q2 — FINAL SUBMISSION COMPLETION ADDENDUM (V5)
# SINGLE GOOGLE COLAB CELL
#
# Run AFTER NHANES_Q2_FINAL_SINGLE_CELL.ipynb in the SAME session.
#
# Outputs:
#  1. Corrected 2007–2016, 100-repeat PSU stability
#  2. Bootstrap duplicate-PSU leakage audit
#  3. Weighted Table 1b by outcome
#  4. Sequential participant-flow diagram
#  5. Updated ROC, PR, calibration, decision curves
#  6. Updated primary/sensitivity AUROC forest plot
#  7. Selected-panel calibration-in-the-large
#  8. Software versions and reporting manifest
#  9. TRIPOD+AI reporting checklist template
#
# Does NOT refit the primary models or optimize on 2017–2018.
# Does NOT claim formal design-based inference.
# =====================================================================

import sys
import json
import platform
import importlib.metadata as md
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.special import expit
from scipy.optimize import brentq

from sklearn.metrics import (
    roc_curve,
    precision_recall_curve,
    roc_auc_score,
    average_precision_score
)

# =====================================================================
# 01 — VERIFY ORIGINAL ANALYSIS
# =====================================================================

REQUIRED_V5 = [
    "OUT", "RAW", "df", "dev", "train", "cal", "test",
    "CLINICAL", "available_labs", "selected_labs",
    "final_predictions", "final_intervals",
    "select_laboratories", "folds_for", "CYCLES",
    "INNER_FOLDS", "audit_labels"
]

missing = [x for x in REQUIRED_V5 if x not in globals()]

if missing:
    raise RuntimeError(
        "First run the complete final NHANES notebook. "
        f"Missing objects: {missing}"
    )

V5 = Path(OUT) / "FINAL_SUBMISSION_V5"
V5_TABLES = V5 / "TABLES"
V5_FIGURES = V5 / "FIGURES"
V5_AUDIT = V5 / "AUDIT"

for folder in [V5, V5_TABLES, V5_FIGURES, V5_AUDIT]:
    folder.mkdir(parents=True, exist_ok=True)

SEED_V5 = 20261030
STABILITY_REPEATS_V5 = 100

def report(msg):
    print(f"[V5] {msg}", flush=True)

def save_table_v5(data, filename):
    data.to_csv(V5_TABLES / filename, index=False)

def save_figure_v5(filename):
    plt.tight_layout()
    plt.savefig(
        V5_FIGURES / filename,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

report("FINAL SUBMISSION COMPLETION STARTED")

# Confirm the analysis population and outcome.
assert set(dev["cycle"].unique()) == {
    "2007-2008", "2009-2010", "2011-2012",
    "2013-2014", "2015-2016"
}

assert set(test["cycle"].unique()) == {"2017-2018"}

assert len(test) == len(final_predictions["full"])

assert train["participant_id"].is_unique
assert test["participant_id"].is_unique

assert set(train["psu_group"]).isdisjoint(
    set(cal["psu_group"])
)

# =====================================================================
# 02 — 100-REPEAT SELECTION STABILITY
# CORRECTED DEVELOPMENT DATA ONLY
# =====================================================================

report("STARTING CORRECTED 2007–2016 STABILITY ANALYSIS")

rng = np.random.default_rng(SEED_V5)

# Original PSU identifiers are preserved.
# Repeated copies of a PSU must share one CV group.

source_groups = {
    str(group_id): part.copy()
    for group_id, part in train.groupby("psu_group")
}

group_ids = np.array(list(source_groups.keys()))

stability_records = []
leakage_records = []

for repeat in range(1, STABILITY_REPEATS_V5 + 1):

    sampled = rng.choice(
        group_ids,
        size=len(group_ids),
        replace=True
    )

    pieces = []

    for draw_number, group_id in enumerate(sampled):

        piece = source_groups[group_id].copy()

        # Important: retain the original PSU grouping.
        piece["psu_group"] = str(group_id)

        # Each draw has an identifier for audit purposes.
        piece["bootstrap_draw"] = draw_number

        pieces.append(piece)

    boot = pd.concat(pieces, ignore_index=True)

    if boot["outcome"].nunique() < 2:
        stability_records.append({
            "repeat": repeat,
            "status": "invalid_class",
            "selected_labs": "",
            "selected_count": np.nan
        })
        continue

    try:
        folds = folds_for(
            boot,
            INNER_FOLDS,
            SEED_V5 + repeat
        )

        leakage_count = 0

        for train_idx, valid_idx in folds:

            train_groups = set(
                boot.iloc[train_idx]["psu_group"]
            )

            valid_groups = set(
                boot.iloc[valid_idx]["psu_group"]
            )

            leakage_count += len(
                train_groups.intersection(valid_groups)
            )

        leakage_records.append({
            "repeat": repeat,
            "duplicate_psu_draws": int(
                len(sampled) - len(set(sampled))
            ),
            "psu_group_leakage": leakage_count,
            "passed": leakage_count == 0
        })

        if leakage_count:
            raise RuntimeError(
                f"PSU leakage detected in repeat {repeat}"
            )

        result = select_laboratories(
            boot,
            CLINICAL,
            available_labs,
            SEED_V5 + repeat
        )

        chosen = list(result["selected"])

        stability_records.append({
            "repeat": repeat,
            "status": "success",
            "selected_labs": ",".join(chosen),
            "selected_count": len(chosen)
        })

    except Exception as exc:

        stability_records.append({
            "repeat": repeat,
            "status": "failed",
            "selected_labs": "",
            "selected_count": np.nan,
            "error": str(exc)
        })

    if repeat % 10 == 0:
        report(f"STABILITY {repeat}/100")

stability_runs_v5 = pd.DataFrame(stability_records)
leakage_audit_v5 = pd.DataFrame(leakage_records)

save_table_v5(
    stability_runs_v5,
    "Supplement_100_Stability_Repeats.csv"
)

save_table_v5(
    leakage_audit_v5,
    "Supplement_PSU_Leakage_Audit.csv"
)

successful = stability_runs_v5.loc[
    stability_runs_v5["status"] == "success"
]

if len(successful) < 90:
    raise RuntimeError(
        f"Only {len(successful)}/100 stability repeats succeeded. "
        "Inspect the failure log."
    )

if not leakage_audit_v5["passed"].all():
    raise RuntimeError("PSU leakage audit failed")

counts = {lab: 0 for lab in available_labs}

for value in successful["selected_labs"]:

    for lab in filter(None, value.split(",")):
        counts[lab] += 1

stability_summary_v5 = pd.DataFrame([
    {
        "laboratory": lab,
        "selected_repeats": counts[lab],
        "successful_repeats": len(successful),
        "selection_frequency": counts[lab] / len(successful)
    }
    for lab in available_labs
]).sort_values(
    "selection_frequency",
    ascending=False
)

save_table_v5(
    stability_summary_v5,
    "Table_Selection_Stability_Corrected.csv"
)

print("\nCORRECTED LABORATORY STABILITY")
display(stability_summary_v5.round(4))

plt.figure(figsize=(9, 6))

plot = stability_summary_v5.sort_values(
    "selection_frequency"
)

plt.barh(
    plot["laboratory"],
    plot["selection_frequency"]
)

plt.xlim(0, 1)
plt.xlabel("Selection frequency")
plt.title("Laboratory selection stability — 100 PSU resamples")

save_figure_v5("Figure_4_Selection_Stability.png")

# =====================================================================
# 03 — TABLE 1b: SURVEY-WEIGHTED BASELINE CHARACTERISTICS
# =====================================================================

report("BUILDING WEIGHTED BASELINE TABLE")

baseline_variables = [
    "age",
    "bmi",
    "waist_circumference",
    "systolic_bp",
    "diastolic_bp",
    "serum_albumin",
    "serum_creatinine",
    "uric_acid",
    "alt",
    "ast",
    "total_protein",
    "total_cholesterol",
    "hdl_cholesterol",
    "hemoglobin",
    "white_blood_cell_count",
    "triglycerides",
    "ggt"
]

def weighted_mean_valid(values, weights):

    valid = (
        np.isfinite(values) &
        np.isfinite(weights) &
        (weights > 0)
    )

    if valid.sum() == 0:
        return np.nan

    return float(
        np.average(
            values[valid],
            weights=weights[valid]
        )
    )

baseline_rows = []

for cycle_group, frame in [
    ("Primary development 2007–2016", dev),
    ("Temporal evaluation 2017–2018", test)
]:

    for outcome_value in [0, 1]:

        part = frame.loc[
            frame["outcome"] == outcome_value
        ]

        weights = part["weight"].to_numpy(dtype=float)

        for variable in baseline_variables:

            if variable not in part.columns:
                continue

            values = pd.to_numeric(
                part[variable],
                errors="coerce"
            ).to_numpy(dtype=float)

            baseline_rows.append({
                "population": cycle_group,
                "outcome": outcome_value,
                "variable": variable,
                "n_nonmissing": int(
                    np.isfinite(values).sum()
                ),
                "weighted_mean": weighted_mean_valid(
                    values, weights
                )
            })

        # Binary / categorical proportions
        binary_variables = [
            "sex",
            "smoking_ever",
            "family_history_diabetes"
        ]

        for variable in binary_variables:

            if variable not in part.columns:
                continue

            values = pd.to_numeric(
                part[variable],
                errors="coerce"
            ).to_numpy(dtype=float)

            if variable == "sex":
                values = np.where(
                    np.isfinite(values),
                    (values == 2).astype(float),
                    np.nan
                )

            baseline_rows.append({
                "population": cycle_group,
                "outcome": outcome_value,
                "variable": (
                    "female" if variable == "sex"
                    else variable
                ),
                "n_nonmissing": int(
                    np.isfinite(values).sum()
                ),
                "weighted_mean": weighted_mean_valid(
                    values, weights
                )
            })

baseline_v5 = pd.DataFrame(baseline_rows)

save_table_v5(
    baseline_v5,
    "Table_1b_Weighted_Baseline_By_Outcome.csv"
)

display(baseline_v5.head(20).round(4))

# =====================================================================
# 04 — SEQUENTIAL PARTICIPANT FLOW
# Reconstruct from the original CDC files, rather than subtracting
# overlapping exclusion counts from the existing cohort summary.
# =====================================================================

report("BUILDING SEQUENTIAL PARTICIPANT FLOW")

import io

flow_rows = []

for cycle, (suffix, year) in CYCLES.items():

    def raw_component(component, columns):

        path = Path(RAW) / f"{component}_{suffix}.XPT"

        if not path.exists():
            raise FileNotFoundError(
                f"Missing original CDC file: {path}"
            )

        raw = pd.read_sas(path, format="xport")
        raw.columns = [str(c).upper() for c in raw.columns]

        present = [
            x for x in columns
            if x in raw.columns
        ]

        return raw[present].copy()

    demo = raw_component(
        "DEMO",
        ["SEQN", "RIDAGEYR", "RIDEXPRG",
         "WTSAF2YR", "SDMVSTRA", "SDMVPSU"]
    )

    diq = raw_component(
        "DIQ",
        ["SEQN", "DIQ010"]
    )

    glu = raw_component(
        "GLU",
        ["SEQN", "LBXGLU", "WTSAF2YR"]
    )

    ghb = raw_component(
        "GHB",
        ["SEQN", "LBXGH"]
    )

    # Avoid duplicate weight columns during merging.
    if "WTSAF2YR" in demo.columns:
        glu = glu.drop(
            columns=["WTSAF2YR"],
            errors="ignore"
        )

    merged = (
        demo
        .merge(diq, on="SEQN", how="left")
        .merge(glu, on="SEQN", how="left")
        .merge(ghb, on="SEQN", how="left")
    )

    steps = [
        ("Initial NHANES participants", np.ones(
            len(merged), dtype=bool
        )),
        ("Age >=20", merged["RIDAGEYR"].ge(20)),
        ("No prior diabetes diagnosis", merged["DIQ010"].eq(2))
    ]

    if "RIDEXPRG" in merged.columns:
        steps.append((
            "Not recorded as pregnant",
            ~merged["RIDEXPRG"].eq(1)
        ))

    steps.extend([
        (
            "HbA1c and fasting glucose observed",
            merged["LBXGH"].notna() &
            merged["LBXGLU"].notna()
        ),
        (
            "Positive fasting weight",
            merged["WTSAF2YR"].gt(0)
        ),
        (
            "Survey design identifiers present",
            merged["SDMVSTRA"].notna() &
            merged["SDMVPSU"].notna()
        )
    ])

    remaining = np.ones(len(merged), dtype=bool)

    for step_name, criterion in steps:

        before = int(remaining.sum())

        remaining &= np.asarray(
            criterion.fillna(False)
            if hasattr(criterion, "fillna")
            else criterion,
            dtype=bool
        )

        after = int(remaining.sum())

        flow_rows.append({
            "cycle": cycle,
            "step": step_name,
            "n_before": before,
            "excluded_at_step": before - after,
            "n_remaining": after
        })

    actual = int(
        (df["cycle"] == cycle).sum()
    )

    if int(remaining.sum()) != actual:
        raise RuntimeError(
            f"Sequential flow mismatch in {cycle}: "
            f"reconstructed={remaining.sum()}, "
            f"final={actual}"
        )

flow_v5 = pd.DataFrame(flow_rows)

save_table_v5(
    flow_v5,
    "Table_Sequential_Exclusions.csv"
)

# Aggregate by role while retaining cycle-level CSV.
flow_v5["role"] = np.where(
    flow_v5["cycle"] == "2017-2018",
    "Temporal evaluation",
    np.where(
        flow_v5["cycle"] == "2005-2006",
        "Sensitivity only",
        "Primary development"
    )
)

flow_summary = (
    flow_v5.groupby(
        ["role", "step"],
        as_index=False,
        sort=False
    )[["excluded_at_step", "n_remaining"]]
    .sum()
)

save_table_v5(
    flow_summary,
    "Table_Sequential_Flow_By_Role.csv"
)

# Draw a compact publication-ready flow diagram.
fig, ax = plt.subplots(figsize=(11, 7))
ax.axis("off")

initial_n = int(
    flow_v5.loc[
        flow_v5["step"] == "Initial NHANES participants",
        "n_before"
    ].sum()
)

flow_text = [
    f"NHANES 2005–2018\nInitial participants: {initial_n:,}"
]

for step_name in flow_v5["step"].drop_duplicates().iloc[1:]:

    excluded = int(
        flow_v5.loc[
            flow_v5["step"] == step_name,
            "excluded_at_step"
        ].sum()
    )

    flow_text.append(
        f"{step_name}\nExcluded at this step: {excluded:,}"
    )

flow_text.append(
    f"Final eligible cohort: {len(df):,}\n"
    f"Primary development 2007–2016: {len(dev):,}\n"
    f"Temporal evaluation 2017–2018: {len(test):,}\n"
    f"2005–2006 sensitivity cohort: "
    f"{int((df['cycle']=='2005-2006').sum()):,}"
)

ys = np.linspace(0.94, 0.06, len(flow_text))

for i, (label, ypos) in enumerate(zip(flow_text, ys)):

    ax.text(
        0.5, ypos, label,
        ha="center", va="center",
        fontsize=9,
        bbox=dict(
            boxstyle="round,pad=0.45",
            facecolor="white",
            edgecolor="black"
        ),
        transform=ax.transAxes
    )

    if i < len(flow_text) - 1:
        ax.annotate(
            "",
            xy=(0.5, ys[i+1] + 0.025),
            xytext=(0.5, ypos - 0.025),
            xycoords=ax.transAxes,
            textcoords=ax.transAxes,
            arrowprops=dict(arrowstyle="->")
        )

save_figure_v5("Figure_1_Participant_Flow.png")

# =====================================================================
# 05 — FINAL PREDICTIONS AND CALIBRATION
# =====================================================================

report("UPDATING TEMPORAL PERFORMANCE FIGURES")

y_v5 = test["outcome"].to_numpy(dtype=int)
w_v5 = test["weight"].to_numpy(dtype=float)

preds_v5 = {
    name: np.asarray(p, dtype=float)
    for name, p in final_predictions.items()
}

def calibration_large_v5(y, p, w):

    p = np.clip(p, 1e-7, 1 - 1e-7)
    z = np.log(p / (1 - p))

    def score(alpha):
        return np.sum(
            w * (y - expit(alpha + z))
        )

    return float(brentq(score, -40, 40))

calibration_rows_v5 = []

for name, p in preds_v5.items():

    calibration_rows_v5.append({
        "model": name,
        "calibration_in_the_large":
            calibration_large_v5(y_v5, p, w_v5),
        "observed_weighted_prevalence":
            float(np.average(y_v5, weights=w_v5)),
        "mean_predicted_risk":
            float(np.average(p, weights=w_v5))
    })

calibration_v5 = pd.DataFrame(calibration_rows_v5)

save_table_v5(
    calibration_v5,
    "Table_Calibration_In_The_Large_All_Models.csv"
)

print("\nCALIBRATION — INCLUDING SELECTED PANEL")
display(calibration_v5.round(5))

# ROC
plt.figure(figsize=(7, 6))

for name, p in preds_v5.items():

    fpr, tpr, _ = roc_curve(
        y_v5, p, sample_weight=w_v5
    )

    auc = roc_auc_score(
        y_v5, p, sample_weight=w_v5
    )

    plt.plot(
        fpr, tpr,
        label=f"{name}: AUROC {auc:.3f}"
    )

plt.plot([0, 1], [0, 1], "--", color="gray")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("Temporal evaluation — ROC")
plt.legend()

save_figure_v5("Figure_2A_ROC.png")

# Precision–recall
plt.figure(figsize=(7, 6))

for name, p in preds_v5.items():

    precision, recall, _ = precision_recall_curve(
        y_v5, p, sample_weight=w_v5
    )

    ap = average_precision_score(
        y_v5, p, sample_weight=w_v5
    )

    plt.plot(
        recall, precision,
        label=f"{name}: AUPRC {ap:.3f}"
    )

prevalence = np.average(y_v5, weights=w_v5)

plt.axhline(
    prevalence, linestyle="--", color="gray"
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Temporal evaluation — Precision–recall")
plt.legend()

save_figure_v5("Figure_2B_Precision_Recall.png")

# Weighted decision curve
thresholds = np.linspace(0.01, 0.15, 100)

def net_benefit_v5(y, p, w, threshold):

    positive = p >= threshold
    total_w = np.sum(w)

    tp = np.sum(w * (positive & (y == 1))) / total_w
    fp = np.sum(w * (positive & (y == 0))) / total_w

    return tp - fp * threshold / (1 - threshold)

plt.figure(figsize=(8, 6))

for name, p in preds_v5.items():

    benefits = [
        net_benefit_v5(y_v5, p, w_v5, t)
        for t in thresholds
    ]

    plt.plot(thresholds, benefits, label=name)

treat_all = [
    prevalence -
    (1 - prevalence) * t / (1 - t)
    for t in thresholds
]

plt.plot(
    thresholds, treat_all,
    "--", label="Treat all"
)

plt.axhline(
    0, color="gray", linestyle=":",
    label="Treat none"
)

plt.xlabel("Risk threshold")
plt.ylabel("Weighted net benefit")
plt.title("Exploratory decision curve analysis")
plt.legend()

save_figure_v5("Figure_2C_Decision_Curves.png")

# Weighted decile calibration
plt.figure(figsize=(7, 6))

calibration_plot_rows = []

for name, p in preds_v5.items():

    order = np.argsort(p)
    groups = np.array_split(order, 10)

    observed = []
    predicted = []

    for group in groups:

        obs = np.average(
            y_v5[group],
            weights=w_v5[group]
        )

        pred = np.average(
            p[group],
            weights=w_v5[group]
        )

        observed.append(obs)
        predicted.append(pred)

        calibration_plot_rows.append({
            "model": name,
            "observed": obs,
            "predicted": pred,
            "n": len(group)
        })

    plt.plot(
        predicted, observed,
        marker="o", label=name
    )

plt.plot(
    [0, 0.3], [0, 0.3],
    "--", color="gray"
)

plt.xlabel("Mean predicted risk")
plt.ylabel("Weighted observed frequency")
plt.title("Temporal calibration")
plt.legend()

save_figure_v5("Figure_3_Calibration.png")

save_table_v5(
    pd.DataFrame(calibration_plot_rows),
    "Supplement_Calibration_Deciles.csv"
)

# =====================================================================
# 06 — UPDATED PRIMARY/SENSITIVITY FOREST PLOT
# =====================================================================

report("UPDATING INCREMENTAL AUROC FOREST PLOT")

forest = final_intervals.loc[
    final_intervals["metric"] == "auroc"
].copy().reset_index(drop=True)

required_periods = {
    "primary_2007_2016",
    "sensitivity_2005_2016"
}

if set(forest["analysis"]) != required_periods:
    raise RuntimeError(
        "Unexpected primary/sensitivity periods"
    )

plt.figure(figsize=(9, 4.5))

for i, row in forest.iterrows():

    center = row["point_difference"]

    plt.errorbar(
        center,
        i,
        xerr=[
            [center - row["lower_95"]],
            [row["upper_95"] - center]
        ],
        fmt="o",
        capsize=5
    )

plt.yticks(
    range(len(forest)),
    forest["analysis"]
)

plt.axvline(0, color="gray", linestyle="--")

plt.xlabel("Incremental AUROC: full minus clinical")
plt.title("Primary and sensitivity analyses")

save_figure_v5(
    "Supplement_Figure_S1_Incremental_AUROC.png"
)

save_table_v5(
    forest,
    "Supplement_Updated_AUROC_Intervals.csv"
)

# =====================================================================
# 07 — GLUCOSE LABEL AND METHOD AUDIT
# =====================================================================

report("EXPORTING GLUCOSE HARMONIZATION AUDIT")

save_table_v5(
    audit_labels,
    "Supplement_Glucose_Label_Changes.csv"
)

glucose_method_audit = pd.DataFrame([
    {
        "cycle": cycle,
        "eligible_n": int((df["cycle"] == cycle).sum()),
        "changed_labels": int(
            df.loc[
                df["cycle"] == cycle,
                "outcome_changed"
            ].sum()
        ),
        "glucose_conversion": (
            "; ".join(
                sorted(
                    df.loc[
                        df["cycle"] == cycle,
                        "glucose_conversion"
                    ].dropna().astype(str).unique()
                )
            )
            if "glucose_conversion" in df.columns
            else "not available"
        ),
        "cdc_equation_independently_verified":
            False,
        "notes": (
            "Verify original instrument, bridging study, "
            "direction and exact coefficients against CDC"
        )
    }
    for cycle in CYCLES
])

save_table_v5(
    glucose_method_audit,
    "Supplement_Glucose_Method_Audit.csv"
)

# Do not automatically assert that a conversion equation
# is valid for every historical instrument.

# =====================================================================
# 08 — SOFTWARE AND REPRODUCIBILITY
# =====================================================================

report("WRITING SOFTWARE MANIFEST")

packages = [
    "numpy",
    "pandas",
    "scipy",
    "scikit-learn",
    "matplotlib",
    "requests"
]

versions = {
    "python": platform.python_version(),
    "platform": platform.platform()
}

for package in packages:
    try:
        versions[package] = md.version(package)
    except Exception:
        versions[package] = "unavailable"

with open(
    V5_AUDIT / "Software_Versions.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        versions, f,
        indent=2,
        ensure_ascii=False
    )

# =====================================================================
# 09 — TRIPOD+AI REPORTING AUDIT TEMPLATE
# =====================================================================

# This is a working audit template, not a claim that the official
# TRIPOD+AI checklist has already been fully satisfied.
# The official item-by-item checklist must be reconciled against
# the final manuscript before submission.

tripod_sections = [
    ("Title and abstract", "Review final title and abstract"),
    ("Background and objectives", "Introduction"),
    ("Data source", "Methods 2.1"),
    ("Participants and eligibility", "Methods 2.2"),
    ("Outcome definition", "Methods 2.3"),
    ("Predictors", "Methods 2.4"),
    ("Sample size and missing data", "Methods and Table 1"),
    ("Data partitions", "Methods 2.5"),
    ("Model development", "Methods 2.6"),
    ("Internal validation", "Methods 2.7"),
    ("Model performance", "Methods 2.8"),
    ("Uncertainty", "Methods 2.9"),
    ("Participant flow", "Figure 1"),
    ("Baseline characteristics", "Table 1b"),
    ("Model specification", "Supplementary material"),
    ("Discrimination", "Tables 2–3"),
    ("Calibration", "Table 2 and Figure 3"),
    ("Clinical utility", "Figure 2C"),
    ("Limitations", "Discussion"),
    ("Ethics and funding", "Declarations"),
    ("Data and code availability", "Declarations")
]

tripod_audit = pd.DataFrame([
    {
        "reporting_topic": topic,
        "proposed_location": location,
        "verified_against_official_TRIPOD_AI": False,
        "final_manuscript_page": "",
        "reviewer_notes": ""
    }
    for topic, location in tripod_sections
])

save_table_v5(
    tripod_audit,
    "TRIPOD_AI_Working_Reporting_Audit.csv"
)

# =====================================================================
# 10 — FINAL READINESS CHECK
# =====================================================================

readiness_v5 = pd.DataFrame([
    {
        "item": "Corrected development stability",
        "status": "completed",
        "detail": f"{len(successful)}/100 successful repeats"
    },
    {
        "item": "Bootstrap PSU leakage",
        "status": "completed",
        "detail": "No original PSU split across folds"
    },
    {
        "item": "Weighted baseline characteristics",
        "status": "completed",
        "detail": "Weighted means and binary proportions"
    },
    {
        "item": "Sequential participant flow",
        "status": "completed",
        "detail": "Reconstructed from original CDC XPT files"
    },
    {
        "item": "Updated publication figures",
        "status": "completed",
        "detail": "ROC, PR, DCA, calibration, stability, AUROC"
    },
    {
        "item": "Selected-panel calibration",
        "status": "completed",
        "detail": "Calibration-in-the-large"
    },
    {
        "item": "Glucose instrument comparability",
        "status": "manual_document_verification_required",
        "detail": "Check exact CDC equations and instrument applicability"
    },
    {
        "item": "Predictor assay comparability",
        "status": "manual_document_verification_required",
        "detail": "Review cycle-specific laboratory documentation"
    },
    {
        "item": "Reference verification",
        "status": "pending",
        "detail": "Verify bibliography and DOI metadata"
    },
    {
        "item": "Official TRIPOD+AI checklist",
        "status": "pending",
        "detail": "Complete official checklist against final pages"
    },
    {
        "item": "Target journal formatting",
        "status": "pending",
        "detail": "Confirm current author instructions"
    }
])

save_table_v5(
    readiness_v5,
    "Final_Submission_Readiness.csv"
)

manifest_v5 = {
    "analysis": "NHANES Q2 submission completion",
    "version": "5.0",
    "created_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "primary_development": "2007–2016",
    "sensitivity_development": "2005–2016",
    "temporal_evaluation": "2017–2018",
    "temporal_test_previously_inspected": True,
    "new_primary_model_fitting": False,
    "new_temporal_hyperparameter_tuning": False,
    "stability_repeats_requested": 100,
    "stability_repeats_successful": len(successful),
    "bootstrap_group_identity": "original PSU preserved",
    "software": versions,
    "glucose_equations_independently_verified": False,
    "predictor_assay_comparability_verified": False,
    "formal_design_based_variance": False,
    "independent_external_validation": False
}

with open(
    V5 / "Final_Submission_Manifest.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        manifest_v5, f,
        indent=2,
        ensure_ascii=False
    )

print("\n" + "=" * 72)
print("FINAL SUBMISSION ADDENDUM COMPLETED")
print("=" * 72)

print("\nSELECTION STABILITY")
display(stability_summary_v5.round(4))

print("\nCALIBRATION")
display(calibration_v5.round(5))

print("\nFINAL READINESS")
display(readiness_v5)

print("\nSOFTWARE")
print(json.dumps(versions, indent=2))

print("\nALL OUTPUTS SAVED TO:")
print(V5)

print("\nIMPORTANT:")
print(
    "Computational reporting is complete. "
    "CDC assay documentation, bibliography, official TRIPOD+AI "
    "checklist and journal requirements still need verification."
)
```

```python
# ================================================================
# NHANES Q2 — COMPLETE RESULTS REPORT FOR CLAUDE
# ONE GOOGLE COLAB CELL
# Reads saved results; does not retrain models.
# ================================================================

import sys, subprocess, json, re
from pathlib import Path
from datetime import datetime

subprocess.check_call([
    sys.executable, "-m", "pip", "-q", "install", "python-docx"
])

import pandas as pd
from docx import Document
from docx.shared import Inches, Pt, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT

try:
    from google.colab import drive, files
    drive.mount("/content/drive")
except ImportError:
    files = None

ROOT = Path(
    "/content/drive/MyDrive/Sta/NHANES_Q2_FINAL_HARMONIZED"
)

if not ROOT.exists():
    raise FileNotFoundError(
        f"Analysis directory not found: {ROOT}"
    )

OUTPUT = ROOT / "NHANES_COMPLETE_RESULTS_FOR_CLAUDE.docx"

doc = Document()
section = doc.sections[0]
section.top_margin = Inches(0.75)
section.bottom_margin = Inches(0.75)
section.left_margin = Inches(0.8)
section.right_margin = Inches(0.8)

normal = doc.styles["Normal"]
normal.font.name = "Calibri"
normal.font.size = Pt(9)

def heading(text, level=1):
    doc.add_heading(text, level=level)

def paragraph(text):
    doc.add_paragraph(str(text))

def add_dataframe(df, title, max_rows=None):
    heading(title, 2)

    if df.empty:
        paragraph("No rows available.")
        return

    df = df.copy()

    # Keep source values unchanged, except formatting for Word.
    df = df.astype(object).where(pd.notnull(df), "")

    if max_rows is not None and len(df) > max_rows:
        shown = df.head(max_rows)
        paragraph(
            f"Displaying {len(shown)} of {len(df)} rows. "
            "Complete data remain in the original CSV."
        )
    else:
        shown = df

    # Split wide tables into readable column blocks.
    columns = list(shown.columns)
    chunks = [
        columns[i:i+7]
        for i in range(0, len(columns), 7)
    ]

    for ci, cols in enumerate(chunks, 1):
        if len(chunks) > 1:
            paragraph(f"Column block {ci}/{len(chunks)}")

        table = doc.add_table(rows=1, cols=len(cols))
        table.style = "Table Grid"
        table.alignment = WD_TABLE_ALIGNMENT.CENTER

        for j, col in enumerate(cols):
            table.rows[0].cells[j].text = str(col)

        for row in shown[cols].itertuples(index=False, name=None):
            cells = table.add_row().cells
            for j, val in enumerate(row):
                if isinstance(val, float):
                    cells[j].text = f"{val:.6f}"
                else:
                    cells[j].text = str(val)[:180]

        doc.add_paragraph("")

def safe_read_csv(path):
    try:
        return pd.read_csv(path, low_memory=False)
    except Exception as exc:
        return pd.DataFrame({
            "Read error": [str(exc)]
        })

def readable_name(path):
    return path.stem.replace("_", " ")

def add_picture(path):
    try:
        heading(readable_name(path), 2)
        doc.add_picture(str(path), width=Inches(6.5))
        doc.paragraphs[-1].alignment = WD_ALIGN_PARAGRAPH.CENTER
        paragraph(f"Source file: {path.relative_to(ROOT)}")
    except Exception as exc:
        paragraph(f"Image insertion failed: {exc}")

# ------------------------------------------------
# COVER
# ------------------------------------------------

title = doc.add_heading(
    "NHANES 2007–2018\nCOMPLETE ANALYSIS RESULTS",
    0
)
title.alignment = WD_ALIGN_PARAGRAPH.CENTER

p = doc.add_paragraph(
    "Comprehensive reproducibility and manuscript preparation report"
)
p.alignment = WD_ALIGN_PARAGRAPH.CENTER

paragraph("Target journal: Statistical Methods in Medical Research (SAGE)")
paragraph(f"Report generated: {datetime.now():%Y-%m-%d %H:%M}")
paragraph(f"Analysis directory: {ROOT}")

heading("Important interpretation notes", 1)

for note in [
    "This document is an export of saved analysis outputs, not an independent verification of their scientific validity.",
    "Primary development: NHANES 2007–2016.",
    "Temporal evaluation: NHANES 2017–2018.",
    "Sensitivity development: NHANES 2005–2016.",
    "Temporal evaluation data were inspected in earlier iterations; do not describe this as untouched external validation.",
    "Reported fixed-prediction PSU bootstrap intervals are approximate, not full survey-design confidence intervals.",
    "CDC glucose conversion coefficients and applicability to historical instruments require independent documentary verification.",
    "Cross-cycle comparability of non-glycemic laboratory assays remains to be verified.",
    "Do not fabricate missing results, references, equations or journal requirements."
]:
    doc.add_paragraph(note, style="List Bullet")

# ------------------------------------------------
# EXECUTIVE RESULTS
# ------------------------------------------------

heading("1. Key final numerical results", 1)

summary = pd.DataFrame([
    ["Eligible participants, all cycles", "14,433"],
    ["Primary development participants", "10,777"],
    ["Temporal evaluation participants", "1,919"],
    ["Temporal evaluation cases", "128"],
    ["Outcome labels changed after harmonization", "62"],
    ["Clinical model AUROC", "0.78126"],
    ["Selected panel AUROC", "0.79879"],
    ["Full laboratory model AUROC", "0.81561"],
    ["Primary full-versus-clinical delta AUROC", "0.03435"],
    ["Primary delta AUROC approximate 95% interval", "0.01635 to 0.05508"],
    ["Sensitivity full-versus-clinical delta AUROC", "0.02229"],
    ["Clinical calibration-in-the-large", "-0.30117"],
    ["Selected calibration-in-the-large", "-0.15538"],
    ["Full calibration-in-the-large", "-0.30712"],
    ["Selection stability successful repetitions", "100/100"]
], columns=["Measure", "Result"])

add_dataframe(summary, "Key results")

# ------------------------------------------------
# FIND ALL SAVED OUTPUTS
# ------------------------------------------------

csv_files = sorted(ROOT.rglob("*.csv"))
image_files = sorted(
    p for p in ROOT.rglob("*")
    if p.suffix.lower() in (".png", ".jpg", ".jpeg")
)
json_files = sorted(ROOT.rglob("*.json"))

# Avoid exporting raw participant-level and replicate-level
# data as enormous Word tables. Include file inventory instead.
raw_terms = [
    "predictions",
    "replicates",
    "participant_level",
    "processed_cohort",
    "harmonized_cohort",
    "raw_data",
    "download_records"
]

def is_large_detail(path):
    name = path.name.lower()
    return any(term in name for term in raw_terms)

priority_keywords = [
    "performance",
    "interval",
    "calibration",
    "stability",
    "baseline",
    "cohort",
    "subgroup",
    "label",
    "exclusion",
    "flow",
    "leakage",
    "readiness",
    "feature",
    "selected",
    "nested",
    "sensitivity"
]

def priority(path):
    name = path.name.lower()
    for i, key in enumerate(priority_keywords):
        if key in name:
            return i
    return 100

csv_files.sort(key=lambda p: (priority(p), str(p)))

heading("2. Full analysis tables", 1)

included = []
indexed_only = []
failed = []

for index, path in enumerate(csv_files, 1):
    relative = str(path.relative_to(ROOT))

    try:
        frame = safe_read_csv(path)

        if (
            is_large_detail(path)
            or len(frame) > 500
            or len(frame.columns) > 30
        ):
            indexed_only.append({
                "file": relative,
                "rows": len(frame),
                "columns": len(frame.columns),
                "reason": "Large/detail-level dataset; indexed rather than reproduced in Word"
            })
            continue

        heading(f"2.{index} {readable_name(path)}", 2)
        paragraph(f"Source: {relative}")
        paragraph(
            f"Dimensions: {len(frame)} rows × {len(frame.columns)} columns"
        )

        # Preserve all rows in small and medium analysis tables.
        add_dataframe(frame, "Results table")

        included.append(relative)

    except Exception as exc:
        failed.append({
            "file": relative,
            "error": str(exc)
        })

# ------------------------------------------------
# FIGURES
# ------------------------------------------------

doc.add_page_break()
heading("3. Publication figures", 1)

# Prefer V5 updated figures where names overlap.
image_files.sort(
    key=lambda p: (
        0 if "FINAL_SUBMISSION_V5" in str(p) else 1,
        str(p)
    )
)

seen_names = set()
included_figures = []

for path in image_files:
    normalized = path.name.lower()

    if normalized in seen_names:
        continue

    seen_names.add(normalized)
    add_picture(path)
    included_figures.append(str(path.relative_to(ROOT)))

# ------------------------------------------------
# MANIFESTS AND SOFTWARE
# ------------------------------------------------

doc.add_page_break()
heading("4. Reproducibility manifests and software", 1)

for path in json_files:
    heading(readable_name(path), 2)
    paragraph(f"Source: {path.relative_to(ROOT)}")

    try:
        content = json.loads(path.read_text(encoding="utf-8"))
        text = json.dumps(
            content,
            indent=2,
            ensure_ascii=False
        )
        paragraph(text[:25000])
        if len(text) > 25000:
            paragraph(
                "Manifest truncated in Word; consult the original JSON."
            )
    except Exception as exc:
        paragraph(f"Unable to read JSON: {exc}")

# ------------------------------------------------
# DATA INVENTORY
# ------------------------------------------------

heading("5. Complete output inventory", 1)

inventory = pd.DataFrame([
    {
        "file": str(path.relative_to(ROOT)),
        "type": path.suffix.lower(),
        "size_kb": round(path.stat().st_size / 1024, 1)
    }
    for path in sorted(ROOT.rglob("*"))
    if path.is_file() and path != OUTPUT
])

add_dataframe(
    inventory,
    "Saved analysis files",
    max_rows=300
)

if indexed_only:
    add_dataframe(
        pd.DataFrame(indexed_only),
        "Large datasets indexed but not embedded"
    )

if failed:
    add_dataframe(
        pd.DataFrame(failed),
        "Files requiring manual inspection"
    )

# ------------------------------------------------
# SCIENTIFIC QUALITY AND MANUSCRIPT HANDOFF
# ------------------------------------------------

doc.add_page_break()
heading("6. Outstanding scientific checks", 1)

checks = [
    (
        "Glucose harmonization",
        "Verify the exact CDC bridging equation, coefficients, direction, applicable cycles, and validity for the 2005–2006 Hitachi 911 instrument."
    ),
    (
        "Predictor assay comparability",
        "Verify changes in non-glycemic laboratory assay methods across cycles and assess whether corrections or limitations are needed."
    ),
    (
        "Table 1b",
        "Ensure the final manuscript table includes both outcome=0 and outcome=1 groups, correct category coding, units and nonmissing sample sizes."
    ),
    (
        "Participant flow",
        "Check that sequential exclusions reproduce the actual eligibility logic, particularly pregnancy and self-reported diabetes criteria."
    ),
    (
        "Software",
        "Distinguish software versions used for original model fitting from those captured during the V5 reporting run."
    ),
    (
        "References",
        "Verify all reference metadata and that each cited paper supports its corresponding statement."
    ),
    (
        "TRIPOD+AI",
        "Complete the official item-by-item checklist and supply final manuscript page references."
    ),
    (
        "Target journal",
        "Confirm current article category, statistical-methodological scope, word limit, reference style and supplementary file requirements."
    )
]

for name, detail in checks:
    heading(name, 2)
    paragraph(detail)

heading("7. Instructions for manuscript preparation", 1)

paragraph(
    "Use the saved numerical results and tables as the source of "
    "truth. Prefer V5 outputs over superseded figures and stability "
    "results. Do not silently reconcile discrepancies or invent "
    "missing values. Clearly separate completed computational "
    "analyses from pending CDC methodological verification."
)

paragraph(
    "The primary research finding is the incremental predictive "
    "value of 12 routine non-glycemic laboratory markers beyond "
    "clinical predictors, assessed with temporal evaluation, "
    "survey-weighted performance metrics, calibration, and "
    "laboratory selection stability."
)

# ------------------------------------------------
# SAVE AND DOWNLOAD
# ------------------------------------------------

doc.save(OUTPUT)

print("=" * 70)
print("WORD REPORT CREATED SUCCESSFULLY")
print("=" * 70)
print("Path:", OUTPUT)
print("CSV tables embedded:", len(included))
print("Figures embedded:", len(included_figures))
print("Large/detail files indexed:", len(indexed_only))
print("Files with read errors:", len(failed))
print("Word file size:", round(OUTPUT.stat().st_size / 1024 / 1024, 2), "MB")

if files is not None:
    files.download(str(OUTPUT))
else:
    print("Download the Word document from the path above.")
```
